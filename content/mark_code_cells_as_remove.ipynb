```python
import nbformat


def set_input_tags(notebook_paths, write=True):
    """
    For each code cell in each notebook:
      - if the cell has 'hide-input', make sure 'remove-input' is NOT present
        (so a cell with both tags ends up with only 'hide-input')
      - otherwise, add 'remove-input'

    Parameters
    ----------
    notebook_paths : iterable of str or pathlib.Path
    write : bool
        If False, nothing is saved (dry run); changes are only reported.
    """
    for path in notebook_paths:
        print(path)
        nb = nbformat.read(path, as_version=4)
        changed = 0

        for cell in nb.cells:
            if cell.cell_type != "code":
                continue

            old_tags = list(cell.metadata.get("tags", []))
            tags = set(old_tags)

            if "hide-input" in tags:
                tags.discard("remove-input")
            else:
                tags.add("remove-input")

            if tags != set(old_tags):
                cell.metadata["tags"] = sorted(tags)
                changed += 1

        if changed and write:
            nbformat.write(nb, path)
        print(f"{path}: {changed} cell(s) updated")
```

```python
import glob
nbs = glob.glob("Lecture*.ipynb")
set_input_tags(nbs)
```
