# Resonance in LC circuits

> In this lecture, we will use the tools we have developed in lecture 3 and 4 to study the behaviour of a circuit that combines both inductors and capacitors. We will find that the combination of a capacitor and an inductor leads to the behaviour of resonance, in which the LC circuit behaves in the same was as a mechanical mass-spring resonator. We will first approach this by looking at the differential equations we generate from Kirchoff's laws, solving them for the step-response behaviour, deriving and providing intuitive explanations of the types of resonances that occur in the circuits. In the second half of the lecture, we will use the tools of complex impedance to derive the frequency response of the circuits, and gain more insight into the resonance phenomena from the cancellation of the reactive impedances of inductors and capacitors at the resonance frequency. Finally, we will discuss the application of these circuits for frequency-selective filtering of electrical signals.

## The equations of motion of RLC circuits

### The series LC circuit 

We will start with a simple circuit consiting an inductor and resistor in series driven by an ideal voltage source:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Ground()
    S = elm.SourceV().up().label(r"$V_s(t)$")
    elm.CurrentLabelInline(direction='out', start=False).at(S).label('$I(t)$')
    elm.Inductor().label("L").right()
    elm.Dot().label("$V_C$")
    elm.Capacitor().down().label("C")
    elm.Line().left()
    display_schematic(d)
```

Using Kirchoff voltage law for the voltage drop around the loop, we will get the following equation:

$$
V_s  = L \frac{dI}{dt} + \frac{Q}{C}
$$

If we take the derivative of this with repect to time, we can see that this is not a *second order* differential equation for the current $I$:

$$
L \frac{d^2I}{dt^2} + \frac{I}{C} = \frac{dV_s}{dt}
$$

As we did in Lecture 2, we will consider a voltage $V_s(t)$ that is a step funciton $\theta(t)$ with height $V_0$:

$$
V_s(t) = V_0 \theta(t) 
$$

```python
mpl.rc('font', size=14)
fig = plt.figure(figsize=(7,3), dpi=100)
plt.axis('off')
plt.plot([-1.2,5.2],[0,0], 'k')
plt.plot([0,0],[-0.5,2.5], 'k')
plt.text(5, -0.4,"$t$")
plt.text(-0.5, 2.5,"$V_s$")
plt.text(-0.5, 1.9, "$V_0$")
t = np.linspace(-1,5, 1000)
V0 = (t>0)*2
plt.plot(t,V0, lw=3, c="blue")
show_plot(fig)
```

For $t>0$, $dV_s / dt = 0$, hence we are interested in the solutions of the following equation:

$$
\frac{d^2I}{dt^2} = -  \frac{I}{LC} 
$$

The most general solution of this equation will be a combination of a cosine and a sine:

$$
I(t) = A \cos(\omega_0 t) + B \sin(\omega_0 t)
$$

In contrast to the solution to the step response of the RC circuit and the RL circuit, which decayed exponentially, the step response of an LC circuit will oscillate in time. The frequency of these oscillations of the undamped LC circuit is known as the [natural frequency](https://en.wikipedia.org/wiki/Natural_frequency) and is given by:

$$
\omega_0 = \frac{1}{\sqrt{LC}}
$$

To find the coefficients $A$ and $B$, we need to apply the boundary conditions at $t=0$. Since the voltage source outputs zero volts for all times $t<0$, the current for $t<0$ must be zero (the capacitor is discharged, there is no current flowing in the inductor). When the voltage step occurs at $t=0$, the circuit will want to suddenly allow current to flow. However, since the inductor will not allow sudden changes in current ($V = L dI/dt$), a "back-EMF" ([counter electromotive force](https://en.wikipedia.org/wiki/Counter-electromotive_force)) will develop across the inductor in response to keep the current zero at $t=0$ and allow it instead to slowly build up. From this, we can applly a boundary condition that $I(t=0) = 0$, leaving us with only the term sine term:

$$
I(t) = B \sin(\omega_0 t)
$$

To find the coefficient B, we can go back to the original equation before we took the derivative:

$$
V_s  = L \frac{dI}{dt} + \frac{Q}{C} 
$$

Since the charge $Q$ on the capacitor starts at zero at $t=0$ and cannot change quickly due to the current being held back by the inductor, at $t=0$, and using that the derivative of a sine is a cosine, we then get: 

$$
V_0 = L \frac{dI}{dt}|_{t=0} = L (B cos(\omega)t) |_{t=0} = L \omega_0 B
$$

which gives:

$$ 
B = \frac{V_0}{\omega_0 L}
$$

and

$$
I(t) = \frac{V_0}{\omega_0 L} \sin(\omega_0 t)
$$

To calculate the voltage $V_C$ across the capacitor, again, we can either integrate the current to get the charge on the capacitor, or we can subtract the voltage across the inductor, giving:

$$
V_C(t) = V_0 (1 - \cos(\omega_0 t))
$$


```python
fig = plt.figure(figsize=(7,6))
plt.subplot(211)
plt.axis('off')
plt.plot([-0.5,3.2],[0,0], 'k')
plt.plot([0,0],[-2.5,2.5], 'k')
plt.text(3.1, -0.45,"$t$")
plt.text(-0.20, 2.5,"$I$")
plt.text(-0.3, 1.7, r"$\frac{V_0}{\omega_0 L}$")
t = np.linspace(-0.3,3, 1000)
V0 = (t>0)*1.8
plt.plot([0,3.2],[V0,V0], lw=1, ls="--", c='k')
plt.plot([1,1],[0,-1.3], lw=1, ls="--", c='k')
plt.text(0.95, -1.8,r"$t = 2 \pi / \omega_0$", size=12)
plt.plot(t,V0*np.sin(t*2*np.pi), c="red", lw=2)

plt.subplot(212)
plt.axis('off')
plt.plot([-0.5,3.2],[0,0], 'k')
plt.plot([0,0],[-0.4,4], 'k')
plt.text(3.1, -0.45,"$t$")
plt.text(-0.3, 3.8,"$V_C$")
plt.text(-0.3, 1.7, r"$V_0$")
t = np.linspace(-0.3,3, 1000)
V0 = (t>0)*1.8
plt.plot([0,3.2],[V0,V0], lw=1, ls="--", c='k')
plt.plot(t,V0*(1-np.cos(t*2*np.pi)), c="blue", lw=2)

show_plot(fig)
```

In the upper plot, we see that the current begins to build up linearly through the inductor, but then reaches a peak value, before oscillating back down again and becoming negative, and then returning back up to zero after a time $t=2\pi / \omega_0$. The maximum height it reaches is $I_{\rm max} = V_0 / \omega_0 L$, which is perhaps not coincidentally closely related to the current one would expect from the impedance of the inductor $V_0 / |Z_L|$.

In the lower plot, we see the voltage on the capacitor. As current starts to flow through the inductor, the capacitor charges up and we build up a voltage on the capacitor. Eventually, the current in the inductor inverts, and the capacitor discharges. What is very strking about the plot is that the voltage on the capacitor at its peak is **actually larger** than the voltage supplied by the source: when charging up, the capacitor **overshoots**! As the capacitor charge climbs up, it reaches the value $V_0$ given by the voltage source, but at that point in time, there is still current flowing in the inductor. To slow down the current flowing in the inductor, you need to build up a negative voltage drop across it to slow the current down again, and hence the voltage of the capacitor gets pushed larger than $V_0$ at the peak when the current is zero. 

Important in this story is that neither the inductor nor the capacitor can dissipate energy: the initial energy added to the circuit by the step response of the voltage source will oscillate back and forevever, first being stored in the magentic fields in the inductor, then being transferred to electric fields in the capacitor as the voltage overshoots to slow the current down, then back into magnetic fields in the inductor as the capacitor discharges again, and then back into the capacitor as the negative overshoot of the current comes back up again. Without any dissipation, this oscillation cycles goes on until $t = \infty$.

This oscillation of energy is exactly analagous to the oscillation of the energy of a mass-spring system between the kinetic energy of the mass and the potential energy stored in the spring. This is also the origin of a common name [tank circuit](https://en.wikipedia.org/wiki/LC_circuit) for LC circuits, based on the analogue of water sloshing around in a tank of water on a ship. Note that even though we have only applied an inpulse step voltage, the resulting currents in the circuit oscillate in time: the voltage step is like a hammer that you have hit the circuit with that makes it "ring" like a bell.

### The series RLC circuit

To damp the oscillations we saw in the series RLC circuit, we will add a resistor to the circuit. This could be an actually intentional resistor, but also in practice, unless the wires of your inductor are superconducting, there will always be a (potentially small) series resistance from the wires in the coil of your inductor. This will lead us to the following circuit:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Ground()
    S = elm.SourceV().up().label(r"$V_s(t)$")
    elm.CurrentLabelInline(direction='out', start=False).at(S).label('I')
    elm.Resistor().label("R").right()
    elm.Inductor().label("L").right()
    elm.Dot().label("$V_C$")
    elm.Capacitor().down().label("C")
    elm.Line().left().length(6)
    display_schematic(d)
```

We will use Kirchoff voltage law for the voltage drop around the loop, giving the following equation:

$$
V_s  = IR + L \frac{dI}{dt} + \frac{Q}{C}
$$

We will again take a derivative to get the following second order differential equation for the current:

$$
L \frac{d^2I}{dt^2} + R\frac{dI}{dt} + \frac{1}{C} I  = \frac{dV_s}{dt}
$$

You will likely recognise this as the equation of motion of the [damped harmonic oscillator](https://en.wikipedia.org/wiki/Harmonic_oscillator#Damped_harmonic_oscillator), with the current playing the role of the position $x$, the inductance the role of the mass $m$, $1/C$ playing the role of the spring constant, and the resistance $R$ playing the role of the [viscous damping coefficient](https://en.wikipedia.org/wiki/Viscous_damping#Single-degree-of-freedom_system) $c$. 

We can solve this by ansatz by showing that the solution is given by: 

$$
I(t) = I_0 e^{-\gamma t} \sin(\omega_1 t + \phi)
$$

Plugging this into the equation above, we can see that it is a solution with a damping coefficient $\gamma$ given by:

$$
\gamma = \frac{R}{2L}
$$

The damping-modified natural frequency $\omega_1$ given by:

$$
\omega_1 = \sqrt{\omega_0^2 - \gamma^2} = \sqrt{\frac{1}{LC} - \left(\frac{R}{2L}\right)^2}
$$

If $\omega_0 > \gamma$, then $\omega_1$ is a real-valued number and the solution will indeed be an oscillating sine function like we assumed in our ansatz. 

At $t=0$, since the circuit starts with no charge and no current, like the series LC circuit, the current at $t=0$ must be zero  $I(t=0) = 0$, giving that $\phi = 0$. Again, in the same way as the series LC circuit, balancing the voltages at $t=0$, we will also find that $I_0 = V_0 / \omega_1 L$, since there is also no voltage drop across the resistor. 

We can also find the voltage $V_C(t)$ across the capacitor:

$$
V_C(t) = V_0 - L\frac{dI}{dt} - RI = V_0 \left[ 1 - e^{-\gamma t} \left( \cos(\omega_1 t) + \frac{\gamma}{\omega_1} \sin(\omega_1 t) \right) \right]
$$

Let's take a look at what these look like for $\gamma = 0.1 \omega_1$:

```python
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure(figsize=(7, 6))

# Normalized parameters for plotting (setting omega_0 = 2*pi so T_0 = 1, V_0 = 1.8 for scaling)
omega_0 = 2 * np.pi*4
gamma = omega_0 / 10
omega_1 = np.sqrt(omega_0**2 - gamma**2)
V0_val = 1.8

# Time array
t = np.linspace(-0.3, 3, 1000)
# Step function for voltage source driving
V_step = (t > 0) * V0_val

# 1. Top subplot: Current I(t)
plt.subplot(211)
plt.axis("off")
plt.plot([-0.5, 3.2], [0, 0], "k")
plt.plot([0, 0], [-2.5, 2.5], "k")
plt.text(3.1, -0.45, "$t$")
plt.text(-0.20, 2.5, "$I$")
plt.plot([2*np.pi/omega_1, 2*np.pi/omega_1], [0, -1.3], lw=1, ls="--", c="k")
plt.text(0.3, -1.8, r"$t = 2 \pi / \omega_1$", size=12)
plt.text(0.3, 1.5, r"$e^{-\gamma t}$", size=12)


# Damped current solution for t > 0 (zero for t <= 0)
# Amplitude factor: V_0 / (omega_1 * L), scaled here relative to V0_val
I_t = np.where(
    t > 0,
    (V0_val / (omega_1 * (V0_val / omega_0)))
    * np.exp(-gamma * t)
    * np.sin(omega_1 * t),
    0,
)
plt.plot(t, I_t*2, c="red", lw=2)
t1 = np.linspace(0,3,1000)
plt.plot(t1, (2*(V0_val / (omega_1 * (V0_val / omega_0))) * np.exp(-gamma * t1))*(t1>0), c="k", ls="--", lw=1)


# 2. Bottom subplot: Capacitor Voltage V_C(t)
plt.subplot(212)
plt.axis("off")
plt.plot([-0.5, 3.2], [0, 0], "k")
plt.plot([0, 0], [-0.4, 4], "k")
plt.text(3.1, -0.45, "$t$")
plt.text(-0.3, 3.8, "$V_C$")
plt.text(-0.3, 1.7, "$V_0$")

plt.plot([0, 3.2], [V0_val, V0_val], lw=1, ls="--", c="k")

# Damped capacitor voltage solution for t > 0 (zero for t <= 0)
VC_t = np.where(
    t > 0,
    V0_val
    * (
        1
        - np.exp(-gamma * t)
        * (np.cos(omega_1 * t) + (gamma / omega_1) * np.sin(omega_1 * t))
    ),
    0,
)
plt.plot(t, VC_t, c="blue", lw=2)

show_plot(fig)
```

Looking at the top plot, we see that like in the series LC circuit, the series RLC circuit also responds to the step in voltage by producing an oscillation in current, but now this oscillation does not go on forever, but is instead damped by an exponential envelope $e^{-\gamma t}$. The damping coefficient in the series RLC circuit is proportional to the resistance:

$$
\gamma = \frac{R}{2L}
$$

The larger the series resistance, the faster the damping occurs. A second thing we can notice about the oscillations in the series RLC circuit is that the frequency of oscillation is slighly lower than that of the natural, undamped frequency: $\omega_1  = \sqrt{\omega_0^2 - \gamma^2}$. 

Looking at the lower plot of $V_C(t)$, we see that it will also experience damped oscillations, but that at long times, we recover back the step in the voltage up to $V_0$: once the oscillations die out, the capacitor does end up charged to a static voltage. We can see, however, that unlike the RC circuit, where there was just an exponential decay up to this new value, once we add the inductor, the voltage can "overshoot" and oscillate before reaching the new value. This "ringing" in response to steps is an indication that somewhere in your circuit, there is a capacitor and inductor that are resonating with each other. If you have sharp enough pulse, and some stray capacitances and inductances in your circuit, you will very frequently see this type of "overshoot" and ringing, an it is a common engineering challenge to get rid of them in high frequency electronics. 

In the plot above, we have chosen a ratio of $\gamma / \omega_0 = 0.1$: the smaller this ratio, the more oscillations you will see before they decay. This is commonly quantified in something called the [quality factor](https://en.wikipedia.org/wiki/Q_factor) defined as:

$$
Q = \frac{\omega_0}{2 \gamma} 
$$

Note that the factor of two here comes from the fact that the quality factor is defined in terms of the energy decay rate: since the energy in the inductor is proportional to $I^2$, the energy decays at a rate $2 \gamma$ based on our convention above. 

For the series RLC, the Q-factor is given by:

$$
Q_{\rm ser} = \frac{1}{R} \sqrt{\frac{L}{C}} = \frac{Z_C}{R}
$$

where $Z_C = \omega_0 L = 1 / (\omega_0 C)$ is a number known as the *characteristic impedance* of the resonator. As we discussed above, for the series RLC circuit, a lower resistance will give a higher Q. The reason for this is that when the circuit is resonating, the resonant current is always forced to flow through the resistor, and so the higher the resistance, the more dissipation, since for a resistor, $P = I^2 R$. The plots above are made with $Q = 1 / 0.1 / 2 = 5$. In the plot you can also "see" about 5 oscillations, which is a "finger in the air" way to estimate the Q-factor from the ringdown plot. For quantitative numbers, you can better fit, or estimate $\gamma$ from the 1/e time of the decay envelope. 

The story above so far with the ringing oscillations holds only if the damping is sufficiently small ($\gamma \ll \omega_0$). If that is not the case, you end up with different solutions: if $\gamma > \omega_0$, then $\omega_1 = \sqrt{\omega_0^2 - \gamma^2}$ becomes imaginary: the sine become a hyperbolic sine and there are no oscillations any more:

```python
import matplotlib.pyplot as plt
import numpy as np

# Parameters
omega_0 = 2 * np.pi
t = np.linspace(-0.3, 3, 1000)

gamma_values = [0.1 * omega_0, 1.0 * omega_0, 10.0 * omega_0]
titles = [
    r"Underdamped ($\gamma = 0.1\omega_0$)",
    r"Critically Damped ($\gamma = \omega_0$)",
    r"Overdamped ($\gamma = 10\omega_0$)",
]

fig, axes = plt.subplots(1, 3, figsize=(10, 3))

for ax, gamma, title in zip(axes, gamma_values, titles):
  plt.sca(ax)
  ax.axis("off")

  # Calculate raw current response
  if gamma < omega_0:
    omega_1 = np.sqrt(omega_0**2 - gamma**2)
    I_t = np.where(t > 0, np.exp(-gamma * t) * np.sin(omega_1 * t), 0)
  elif np.isclose(gamma, omega_0):
    I_t = np.where(t > 0, t * np.exp(-gamma * t), 0)
  else:
    s1 = -gamma + np.sqrt(gamma**2 - omega_0**2)
    s2 = -gamma - np.sqrt(gamma**2 - omega_0**2)
    I_t = np.where(t > 0, np.exp(s1 * t) - np.exp(s2 * t), 0)

  # Normalize each curve independently so it fills the same visual height
  max_val = np.max(np.abs(I_t))
  if max_val > 0:
    I_t_normalized = I_t / max_val
  else:
    I_t_normalized = I_t

  # Plot axes centered symmetrically (y from -1.5 to 1.5)
  ax.plot([-0.5, 3.2], [0, 0], "k")
  ax.plot([0, 0], [-1.5, 1.5], "k")
  ax.set_ylim(-1.5, 1.5)

  ax.text(3.1, -0.15, "$t$")
  ax.text(-0.20, 1.4, "$I$")
  ax.set_title(title, fontsize=12, pad=15)

  ax.plot(t, I_t_normalized, c="red", lw=2)

plt.tight_layout()
show_plot(fig)
```

In the underdamped limit, the current oscillates as we discussed above. For critical damping, corresponding to $\gamma = \omega_0$, there is exactly one bump followed by a decay. In the overdamped limit $\gamma >> \omega_0$, there is first a fast rise in the current with a time constant $L/R$ associated with the energisation of the inductor, followed by a long decay with a time constant $RC$ associated with the charging of the capacitor. The overdamped regime looks a lot like the RC circuit we saw in lecture 3, except that in the theoretical ideal RC circuit model, the jump in the current at $t=0$ is really an instantaneous step function, while here with the inductor is it now smoothed out by the $L/R$ time constant.

### The parallel RLC

We can also combine the resistor, inductor, and capacitor in parallel instead of in series. Specifically, we will consider a parallel RLC circuit driven by an ideal current source that provides a step in the current from 0 to $I_0$ at $t = 0$:


```python
with schemdraw.Drawing(show=False) as d:
    elm.Ground()
    S = elm.SourceI().up().label(r"$I_s(t)$")
    elm.Line().right()
    d.push()
    elm.Resistor().label("R").down()
    d.pop()
    elm.Line().right().length(2)
    d.push()
    elm.Inductor().label("L").down()
    d.pop()
    elm.Line().right().length(2)
    elm.Dot().label("$V(t)$")
    elm.Capacitor().label("C").down()
    d.pop()
    elm.Line().left().length(7)
    display_schematic(d)
```

And now we will ask the question: what is the solution for V(t)? 

We will leave it as a homework exercise to show that the solution in the underdamped regime is given by:

$$
V(t) = \frac{I_0}{\omega_1 C} e^{-\gamma t} \sin(\omega_1 t)
$$

Note that the voltage now oscillates to zero and there is no step: at long times, the inductor short circuits the resistor and capacitor giving no DC output voltage. The damping-modified oscillation frequency is once again given by: 

$$
\omega_1 = \sqrt{\omega_0^2 - \gamma^2}
$$

For the parallel RLC, though, we will have a damping rate:

$$
\gamma_{\rm par} = \frac{1}{2RC}
$$

If you look back, you will remember that the resitance $R$ was in the numerator for the series RLC circuit: bigger resistance gave more damping. Here, the resistance is in the denominator: bigger resistance gives *less* damping! While that may seem at first counter intuitive, if you think about how currents will flow in the parallel RLC, then you can understand what is going on: if the resistance is smaller in the parallel RLC, then *more* current will flow through the resistor, given *more* damping. If the resitance is infinite, then *no* current flows through the resistor: all currents flow only the through the non-dissipative inductor and capacitor, and there is no damping at all. 

This also comes back if you look at the formula for the Q-factor of the parallel RLC circuit: 

$$
Q_{\rm par} = \frac{R}{Z_C} = \frac{1}{Q_{\rm ser}}
$$

Again, you see here the opposite behaviour: the parallel RLC circuit has a high quality factor for a large resistance, while exactly the opposite is true for the series RLC circuit. 

## The impedance and frequency response RLC circuits

We will now switch gears and take a look at the response of (R)LC circuits to an oscillating driving voltage or current. We will do this, of course, using the framework of complex impedance. 

### Coherent cancellation of currents in parallel and series LC circuits

We will first consider a series combination of an inductor and a capacitor and determine the effective combined impedance of the circuit:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Inductor().label("L").right().at((0,0))
    elm.Capacitor().right().label("C")
    elm.ResistorIEC().label(r"$Z_{\rm eq}$").right().at((9,0))
    elm.Arrow().length(1).at((7,0))
    display_schematic(d)
    
```

Impedances sum in series and so the equivalent impedance of this series combination is:

$$
Z_{\rm eq,ser} = j \omega L + \frac{1}{j \omega C} 
= j \omega L \left(1 - \frac{1}{\omega^2 L C}\right)
= j \omega L \left(1 - \frac{\omega_0^2}{\omega^2}\right)
$$

where $\omega_0$ is the natural frequency again. Since we have only non-dissipative elements, the impedance is purely imaginary valued. It is useful to plot ${\rm Im}(Z_{\rm eq})$ to see what it looks like:

```python
fig = plt.figure(figsize=(6,4))
plt.axis("off")
xlim = (-0.5,3)
ylim = (-1,1)
plt.plot(xlim, [0, 0], "k") #x axis
plt.plot([0, 0], ylim, "k") #y axis
plt.xlim(xlim)
plt.ylim(ylim)
plt.text(0, ylim[1]*1.1, r"${\rm Im}(Z_{\rm eq, ser})$", ha='center')
plt.text(xlim[1]*1.05, 0, r"$\omega / \omega_0$", va='center')

plt.plot((1,1),(0.2,-0.2), "k:")
plt.text(1.3,-0.3, r"$\omega = \omega_0$", ha="center")


x = np.linspace(0.1,xlim[1],100)
plt.plot(x, (1-1/x**2), c = "blue")
show_plot(fig)
```

At low frequencies, the impedance is dominated by the $1/\omega$ dependence of the capacitor, while at high frequencies, the linear scaling of the impedance with frequency $\omega$ from the inductor takes over. 

A super imporant point of this plot is the difference in **sign** of the impedance of the inductor and capacitor: the capacitor contributes a negative imaginary impedance, while the inductor contributes a positive imaginary impedance. 

This difference in sigh leads to a very special point at $\omega = \omega_0$: at the natural resonance frequency, the two impedences **cancel**! Even though both the capacitor and inductor on their own have non-zero impedance at $\omega_0$, when you put them in series, these two **exactly cancel** and the series combination goes to zero!!! The circuit looks like a short circuit at exactly one frequency where the magnitude of the impedance vanishes:

```python
fig = plt.figure(figsize=(6,4))
plt.axis("off")
xlim = (-0.5,3)
ylim = (-0.2,1)
plt.plot(xlim, [0, 0], "k") #x axis
plt.plot([0, 0], ylim, "k") #y axis
plt.xlim(xlim)
plt.ylim(ylim)
plt.text(0, ylim[1]*1.1, r"$|Z_{\rm eq,ser}|$", ha='center')
plt.text(xlim[1]*1.05, 0, r"$\omega / \omega_0$", va='center')

plt.plot((1,1),(0,-0.1), "k:")
plt.text(1,-0.2, r"$\omega = \omega_0$", ha="center")

x = np.linspace(0.1,xlim[1],1000)
plt.plot(x, np.abs((1-1/x**2)), c = "magenta")
show_plot(fig)
```

The series LC circuit is has a very strange behaviour that it will resist the flow of current at all frequencies, except exactly at the resonance frequency, where it will allow infinite current to flow for any applied voltage.

We can also look at what happens when we put the inductor and capacitor in parallel:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Line().length(1).right().at((0,0))
    d.push()
    elm.Line().length(0.5).up()
    elm.Inductor().label("L").right()
    elm.Line().length(0.5).down()
    elm.Line().length(1).right()
    d.pop()
    elm.Line().length(0.5).down()
    elm.Capacitor().right().label("C", loc="bottom")
    elm.Line().length(0.5).up()
    
    elm.Arrow().right().length(1).at((6,0))

    elm.ResistorIEC().label(r"$Z_{\rm eq}$").right().at((8,0))
    
    display_schematic(d)  
```

Parallel impedances add in inverse, giving:

$$
\frac{1}{Z_{\rm eq, par}} = j \omega C + \frac{1}{j \omega L} = \frac{1- \omega^2 LC}{j \omega L}
$$

Inverting this, we get:

$$
Z_{\rm eq, par} = \frac{j \omega L}{1- \omega^2 LC}
$$

In constrast the vanishing impedance of the series LC circuit at resonance, the impedance of the parallel LC circuit diverges at resonance:

```python
fig = plt.figure(figsize=(12,4))

plt.subplot(121)
plt.axis("off")
xlim = (-0.5,3)
ylim = (-10,10)
plt.plot(xlim, [0, 0], "k") #x axis
plt.plot([0, 0], ylim, "k") #y axis
plt.xlim(xlim)
plt.ylim(ylim)
plt.text(0, ylim[1]*1.1, r"${\rm Im}(Z_{\rm eq, par})$", ha='center')
plt.text(xlim[1]*1.05, 0, r"$\omega / \omega_0$", va='center')
plt.plot((1,1),(-10,10), "k:")

x1 = np.linspace(0.1,0.999,100)
x2 = np.linspace(1.001,xlim[1],100)
plt.plot(x1, 1/(1-1/x1**2), c = "blue")
plt.plot(x2, 1/(1-1/x2**2), c = "blue")

plt.subplot(122)
plt.axis("off")
xlim = (-0.5,3)
ylim = (-1,10)
plt.plot(xlim, [0, 0], "k") #x axis
plt.plot([0, 0], ylim, "k") #y axis
plt.xlim(xlim)
plt.ylim(ylim)
plt.text(0, ylim[1]*1.1, r"$|Z_{\rm eq,par}|$", ha='center')
plt.text(xlim[1]*1.05, 0, r"$\omega / \omega_0$", va='center')
plt.plot((1,1),(-1,10), "k:")

x1 = np.linspace(0.1,0.999,100)
x2 = np.linspace(1.001,xlim[1],100)
plt.plot(x1, -1/(1-1/x1**2), c = "magenta")
plt.plot(x2, 1/(1-1/x2**2), c = "magenta")

show_plot(fig)
```

Away from resonance, the series LC allows current to flow, but at resonance with $\omega = \omega_0$, the currents flowing through the capacitor and inductor exactly cancel each other and all current is blocked: at this frequency, the parallel combination looks like an open circuit with infinite impedance.

### Series and parallel RLC circuits

#### Series RLC circuit

Now let's go back to real life and start adding some resistances to our circuit. We will start with the series RLC circuit and consider driving it with an AC voltage source:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Ground()
    S = elm.SourceSin().up().label(r"$V_s(\omega)$")
    elm.CurrentLabelInline(direction='out', start=False).at(S).label(r'$I(\omega)$')
    elm.Resistor().label("R").right()
    elm.Dot().label(r"$V_{LC}$")
    elm.Inductor().label("L").right()
    elm.Dot().label("$V_C$")
    elm.Capacitor().down().label("C")
    elm.Line().left().length(6)
    display_schematic(d)
```

As discussed earlier, even if you do not intentionally add a resistor, there will always be a series resistance in the circuit from (i) the output impedance of the voltage source and (ii) the series resistance of the coil. 

For the AC response, we will consider $V_s(t) = V_0 e^{j\omega t}$ (in phasor notaion, as you notice). The current is then given by:

$$
I = \frac{V_0}{R + j\omega L + \frac{1}{j\omega C}} 
$$

Using $\omega_0 = 1/\sqrt{LC}$ and $\gamma = R / 2L$, we can rewrite this as:

$$
I = \frac{V_0}{L}\  \frac{\omega}{2\gamma\omega + j(\omega^2 - \omega_0^2)}
$$

Let's have a look at what this looks like:

```python
import matplotlib.pyplot as plt
import numpy as np

# Damping in units of Gamma, this is a Q of 5
gamma = 0.05

# Frequency range normalized: omega / omega_0 from 10^-3 to 10^3
omega = np.linspace(1e-3, 3, 1000)
I = omega / (2*gamma*omega + 1j*(omega**2-1))
I_magnitude = np.abs(I)
I_magnitude /= np.max(I_magnitude)
phi_deg = np.angle(I)/np.pi*180

# Create side-by-side subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Magnitude (Amplitude) Bode Plot
ax1.plot(omega, I_magnitude, color="b", linewidth=2)
ax1.set_xlabel(r"$\omega / \omega_0$")
ax1.set_ylabel(r"$|I(\omega)|\ /\ |I(\omega_0)|$")
ax1.grid(True, which="both", linestyle="--", alpha=0.7)

# Phase Bode Plot
ax2.plot(omega, phi_deg, color="r", linewidth=2)
ax2.set_xlabel(r"$\omega / \omega_0$")
ax2.set_ylabel(r"$\angle I(\omega)$")
#ax2.set_ylim(-5, 9)
ax2.set_yticks(np.arange(-90, 91, 45))  # Sets y-ticks in increments of 15 degrees from 0 to 90
ax2.grid(True, which="both", linestyle="--", alpha=0.7)

plt.tight_layout()
show_plot(fig)
```

Looking at the magintude of the current in the left plot, this agrees with the intuition we developed above with the series LC: at resonance, the impedance of the inductor and the capacitor cancel, creating a peak in the current. Because we now have included a series resistor, this peak no longer goes to infinity but instead takes on a value we can evaluate by taking $\omega = \omega_0$ in the forumla above:

$$
I(\omega_0) 
= \frac{V_0}{L} \frac{1}{2\gamma} 
= \frac{V_0}{L} \frac{2L}{2R} 
= \frac{V_0}{R}
$$

We see here now also from the math that at resonance, it is only the resistor that limits the current. 

In the right plot, we can see the phase of the current, which starts at 90 degree and then swings by 180 degrees to -90 degrees. We can also understand this intuitively: at low frequencies, the capacitor dominates the impedance, giving a +90 degree phase angle, while at high frequencies, the inductor dominates, giving the opposite phase angle ($j$ is in the denominator for a capacitor). 

We can also plot the voltage across the capacitor, given by: 

$$
V_C(\omega) = Z_C I(\omega)
= \frac{I(\omega)}{j \omega C}
= \frac{V_0}{CL}\  \frac{1}{j2\gamma\omega + (\omega_0^2 - \omega^2)}
= V_0 \ \frac{\omega_0^2}{j2\gamma\omega + (\omega_0^2 - \omega^2)}
$$

Let's take a look at this for $\gamma = \omega_0 / 20$:

```python
import matplotlib.pyplot as plt
import numpy as np

# Damping in units of Gamma, this is a Q of 5
gamma = 0.05

# Frequency range normalized: omega / omega_0 from 10^-3 to 10^3
omega = np.linspace(1e-3, 3, 1000)
I = 1 / (1j*2*gamma*omega + (1-omega**2))
I_magnitude = np.abs(I)
phi_deg = np.angle(I)/np.pi*180

# Create side-by-side subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Magnitude (Amplitude) Bode Plot
ax1.plot(omega, I_magnitude, color="b", linewidth=2)
ax1.set_xlabel(r"$\omega / \omega_0$")
ax1.set_ylabel(r"$|V_C(\omega)|\ /\ |V_0|$")
ax1.grid(True, which="both", linestyle="--", alpha=0.7)

# Phase Bode Plot
ax2.plot(omega, phi_deg, color="r", linewidth=2)
ax2.set_xlabel(r"$\omega / \omega_0$")
ax2.set_ylabel(r"$\angle V_C(\omega)$")
ax2.set_ylim(-190, 10)
ax2.set_yticks(np.arange(-180, 1, 45))  # Sets y-ticks in increments of 15 degrees from 0 to 90
ax2.grid(True, which="both", linestyle="--", alpha=0.7)

plt.tight_layout()
show_plot(fig)
```

At low frequencies, the capacitor blocks the current and the voltage at the input is just directly transferred to the output. 

At resonance, something a bit non-intuitive happens: the output voltage is *larger* than the input voltage! This is due to resonance: at resonance, to match the opposing and cancelling currents in the inductor and the capacitor, the voltage of the capacitor needs to swing to bigger voltages than the initial input voltage. This is analogous to a mass on a spring: if you apply a force at resonance, the resulting amplitude is larger than if you applied a static force of the same amplitude. 

We can also look at the voltage ratio at resonance:

$$
\frac{V_C(\omega_0)}{V_0} = \frac{\omega_0}{2\gamma} = Q
$$

At resonance, the output voltage at the capacitor is amplified by a factor given by Q. 

Note that although the output voltage is amplified, making this a resonant "voltage amplifier", this is not an active circuit: there no battery supplying energy, and so it cannot amplify energy; it is not a "power amplifier". If you work out the equivalent output impedance of the node connected to the capacitor at resonance, which is a good exercise for your homework, you will find that it is much higher than the output impedance at DC. From this, you can show that although the voltage is amplified at resonance, it's ability to transfer power to a load resistor is not larger, which makes sense since the passive circuit does not have a source of energy. This will be different when we look at active transistor amplifiers later in the course. 

Like the current, the phase angle of the voltage also swings by 180 degrees as we go through resonance, reaching 90 degrees at resonance, although it does so starting from zero degrees: at DC, where no current flows through the capacitor, the output voltage is in phase with the input voltage.

If you think if this as a filter, with $V_0$ as the input and $V_C$ as the output, then this can also be seen as a [band pass filter](https://en.wikipedia.org/wiki/Band-pass_filter) that efficiently lets one frequency through and blocks others. 

Finally, we can also look at the voltage right after the resistor, which drops across *both* the inductor and capacitor (see the node in the circuit diagram above):

$$
V_{LC}(\omega) = V_0 - I(\omega)R = \frac{\omega^2 -\omega^2}{(\omega_0^2 - \omega^2) + j 2\gamma \omega} \tilde{V}_0
$$

```python
import matplotlib.pyplot as plt
import numpy as np

# Damping in units of Gamma, this is a Q of 5
gamma = 0.05

# Frequency range normalized: omega / omega_0 from 10^-3 to 10^3
omega = np.linspace(0, 3, 3001)
I = (omega**2-1) / (1j*2*gamma*omega + (1-omega**2))
I_magnitude = np.abs(I)
phi_deg = np.angle(I)/np.pi*180

# Create side-by-side subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Magnitude (Amplitude) Bode Plot
ax1.plot(omega, I_magnitude, color="b", linewidth=2)
ax1.set_xlabel(r"$\omega / \omega_0$")
ax1.set_ylabel(r"$|V_C(\omega)|\ /\ |V_0|$")
ax1.grid(True, which="both", linestyle="--", alpha=0.7)

# Phase Bode Plot
ax2.plot(omega, phi_deg, color="r", linewidth=2)
ax2.set_xlabel(r"$\omega / \omega_0$")
ax2.set_ylabel(r"$\angle V_C(\omega)$")
ax2.set_ylim(-190, 190)
ax2.set_yticks(np.arange(-180, 181, 45))  # Sets y-ticks in increments of 15 degrees from 0 to 90
ax2.grid(True, which="both", linestyle="--", alpha=0.7)

plt.tight_layout()
show_plot(fig)
```

Thinking about this as a filter, we can see that if we take the output across both the L and the C, then this circuit now looks like a [notch filter](https://en.wikipedia.org/wiki/Band-stop_filter) that blocks a specific frequency. We can also understand this behaviour from the series impedance of the LC: at resonance, the series LC looks like a zero-ohm short circuit to ground due to the cancellation of the inductive and capacitive impedances. At resonance, all voltage falls across the resistor and the output of the circuit at this node is zero. 

If we look at the phase, we do see a bit of a funny jump of 180 degress at resonance. We can understand this better in a [polar plot](https://en.wikipedia.org/wiki/Polar_coordinate_system#Complex_numbers) of the complex plane where we plot the real and imaginary parts of the output voltage phasor:

```python
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure(figsize=(4, 4))
ax = plt.gca()

I = I[0:int(len(I)*0.6)]
# 1. Plot the main curve
plt.plot(I.real, I.imag, color="magenta", lw=2)
plt.xlim(-1.1, 1.1)
plt.ylim(-1.1, 1.1)
plt.yticks([-1, 0, 1])
plt.xticks([-1, 0, 1])
plt.grid(linestyle="--", alpha=0.7)
plt.ylabel(r"Im($V_{LC}(\omega)$) / $V_0$")
plt.xlabel(r"Re($V_{LC}(\omega)$) / $V_0$")

# 2. Third plot item as a filled black circle at (0, 0)
plt.plot(0, 0, marker="o", color="magenta", markersize=6)

# 3. Circular arc centered at (-0.5, 0) with clockwise rotation
center_x, center_y = -0.5, 0.0
r = 0.58  # Radius just outside the circle

# Arc span from 120° down to 65° (leaving 65° to 60° for the arrowhead)
theta_arc = np.linspace(np.deg2rad(120), np.deg2rad(65), 100)
x_arc = center_x + r * np.cos(theta_arc)
y_arc = center_y + r * np.sin(theta_arc)
plt.plot(x_arc, y_arc, color="black", lw=1.5)
plt.title("Polar resonance plot", fontsize=14, pad=12)

# Arrowhead segment from 65° to 60° ensuring perfect alignment
rad_start = np.deg2rad(65)
rad_end = np.deg2rad(60)

x_start = center_x + r * np.cos(rad_start)
y_start = center_y + r * np.sin(rad_start)
x_end = center_x + r * np.cos(rad_end)
y_end = center_y + r * np.sin(rad_end)

ax.annotate(
    "",
    xy=(x_end, y_end),
    xytext=(x_start, y_start),
    arrowprops=dict(
        arrowstyle="->",
        lw=1.5,
        color="black",
        shrinkA=0,
        shrinkB=0,
    ),
)

plt.text(-0.5,0.7, r"Increasing $\omega$", fontsize=12, ha='center') 
plt.text(0.1,0.0, r"$\omega = \omega_0$", fontsize=12, va = "center", c='magenta') 

show_plot(fig)
```

In this plot, we plot the data from the last plot but in a different way: we plot the imaginary part of the voltage phasor on the y-axis and the real part on the x-axis. The curve starts on the left on the x-axis: this corresponds to zero frequency. As we increase the frequency, the points trace out a circle on this complex polar plane that rotate in clockwise direction (using the convention of phase defintion we have used). At resonance, the curve passes through the x-axis, then as you past resonance, it swings down and eventually comes back to the x-axis as $\omega \rightarrow \infty$. 

From this plot, we can now also see why there is a jump in the phase in the plots above: as we pass through resonance, the curve touches the origin of the plot exactly and the angle with respect to the origin jumps from 90 degrees to -90 degrees.

### Linewidth of resonances and the Lorentzian approximation

As we saw above, the frequency response of damped LC circuits will typically involve a sharp feature in frequency at resonance, either a narrow peak or a narrow dip. 

How sharp it the resonance in frequency? Before we try to answer this quantitatively, we will first address a common approximation of resonance lineshapes for $Q \gg 1$. In this limit, we can simplify the expressions. In particular, if we are interested only in frequencies close to the resonance, the term in the denominator $\omega^2 - \omega_0^2$ can be simplified as:

$$
\omega^2 - \omega_0^2 = (\omega_0 + \omega)(\omega_0 - \omega) \approx 2\omega_0 (\omega_0 - \omega)
$$

Using $2j\omega\gamma \approx 2j\gamma \omega_0$, this commonly occuring denominator we saw many times in the expressions above can then be simplified to:

$$
\frac{1}{(\omega_0^2 - \omega^2) + j 2\gamma \omega}
\approx \left(\frac{1}{2j \omega_0}\right) \frac{1}{\gamma + j(\omega - \omega_0)}
$$

The second term in this expression is known as a complex representation of a [Lorentzian lineshape](https://en.wikipedia.org/wiki/Spectral_line_shape#Lorentzian) $L(\omega)$:

$$
L(\omega) = \frac{1}{\gamma + j (\omega - \omega_0)}
$$

Let's take a look at what it looks like and how good the Lorentzian approximation is:

```python
# Create side-by-side subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

Q = 2

for Q in 2,5,10:
    w = np.linspace(0,2,1000)
    gam = 1/2/Q
    ho = 1/(1-w**2 + 2j*w*gam)
    lor = 1/2j/(gam + 1j*(w-1))
    
    ax1.axvline(0, ls=":", c="gray")
    ax1.axhline(0, ls=":", c="gray")
    
    (line,) = ax1.plot(w, np.abs(ho), label=f"Q = {Q}")
    ax1.plot(w, np.abs(lor), color=line.get_color(), ls="--")
    ax1.set_xlabel(r"$\omega/\omega_0$")
    ax1.set_ylabel(r"$|L(\omega)|$")
    ax1.legend(fontsize=10)

for Q in 100,300,1000:
    w = np.linspace(0.99,1.01,1000)
    gam = 1/2/Q
    ho = 1/(1-w**2 + 2j*w*gam)
    lor = 1/2j/(gam + 1j*(w-1))
    
    (line,) = ax2.semilogy(w, np.abs(ho), label=f"Q = {Q}")
    ax2.semilogy(w, np.abs(lor), color=line.get_color(), ls="--", lw=2)
    ax2.set_xlabel(r"$\omega/\omega_0$")
    ax2.set_ylabel(r"$|L(\omega)|$")
    ax2.legend(fontsize=10)

plt.tight_layout()
show_plot(fig)
```

Already for Q = 10, the Lorentzian is a good approximation near resonance, and for higher Q, near resonance the two curves lie directly on top of each other. 

The idea that "higher Q gives a narrower peak" can be captured by calculating the [linewidth](https://en.wikipedia.org/wiki/Spectral_line_shape#Lorentzian) of the resonance, defined by the full width at half maximum of the intensity (power). In our case, because power is proportional to $|V|^2$, we look at the full width at half maximum of $|L(\omega)|^2$:

$$
{\rm Linewidth}\ \equiv\  {\rm FWHM\ of\ } |L(\omega)|^2 = 2\gamma = \frac{\omega}{Q}
$$

Here we see now explicitly mathematically that the linewidth of the resonance is inversely proportional to $Q$: the higher the quality factor, the longer it takes the oscillations to decay when you "strike" the LC circuit with a pulse, and the sharper the resonance is of the circuit when you drive it with an oscillating current or voltage. 

## Applications of LC resonant circuits

OK, we've dug pretty deep into the physics of resonance in LC circuits and their behaviour. But you might be now thinking: what are they useful for? 

LC circuit found their first application in the radios of early days, basically being reverse engineered in early [spark gap](https://en.wikipedia.org/wiki/Spark-gap_transmitter) transmitter radios which enabled the first transmission via electromagnetic waves based on predictions from Maxwell's equations.

  * **Transmitters and Receivers.** Used as frequency-selective networks to [tune](https://en.wikipedia.org/wiki/LC_circuit#Applications) into specific carrier frequencies for radio, television, and wireless communication 
* **Filtering and Signal Processing**
  * **Band-Pass Filters.** Series LC circuits offer very low impedance at resonance, letting a target frequency pass through while blocking others
  * **Band-Stop / Notch Filters.** Parallel LC circuits offer very high impedance at resonance, blocking a specific frequency while letting others pass 
* **Oscillators and Signal Generation**
  * **Tone and Clock Generators.** Paired with an active amplifier (such as in Hartley or Colpitts topologies) in a positive feedback loop to generate continuous sinusoidal AC signals, radio waves, or timing pulses. Also commonly used for this are piezo crystal resonators, the famous [quartz crystal oscillators](https://en.wikipedia.org/wiki/Crystal_oscillator) that enabled highly accurate digital watches. Although they are mechanical resonances, the piezoelectric coupling
* **Induction Heating and Power Systems**
  * **Induction Heating.** Parallel and series LC tank circuits in an [induction heating](https://en.wikipedia.org/wiki/Induction_heating) system pass high circulating currents to heat things by inductive power transfer, common in modern cooktops, replacing gas.
  * **Wireless Power Transfer (WPT) and near-field communiction (NFC).** Magnetically coupled resonant LC circuits enable efficient medium-range energy transfer for sensors, bioelectronics, and wireless charging devices using [wireless power transfer](https://en.wikipedia.org/wiki/Wireless_power_transfer). Your phone can use this for charging, your bank pass uses this for powering the chip that sends your bank information to the terminal, and your phone uses it for NFC communications when you pay in the supermarket. [Near-filed communication](https://en.wikipedia.org/wiki/Near-field_communication#Design) is also based on inductive signal transfer between resonant LC circuit build into your phone and the terminal.

Finally, in quantum research, the resonance of the LC circuit is the basic building block of the [superconducting quantum computer](https://en.wikipedia.org/wiki/Superconducting_quantum_computing), which consists of thousands of superconducting LC circuits build on a chip from superconducting metal that resonate at GHz frequencies but that can behave in a quantum way, such as being in a superposition having currents flowing through an inductor simultaneously in two directions at the same time.
