## The equations of motion of RLC circuits

### The series LC circuit 

We will start with a simple circuit consiting an inductor and resistor in series driven by an ideal voltage source:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Ground()
    S = elm.SourceV().up().label(r"$V_s(t)$")
    elm.CurrentLabelInline(direction='out', start=False).at(S).label('I')
    elm.Inductor().label("L").right()
    elm.Dot().label("$V_C$")
    elm.Capacitor().down().label("C")
    elm.Line().left()
    display_schematic(d)
```

Using Kirchoff voltage law for the voltage drop around the loop, we will get the following equation:

$$
V_s  = L \frac{dI}{dt} + \frac{Q}{C}
$$

If we take the derivative of this with repect to time, we can see that this is not a *second order* differential equation for the current $I$:

$$
L \frac{d^2I}{dt^2} + \frac{I}{C} = \frac{dV_s}{dt}
$$

As we did in Lecture 2, we will consider a voltage $V_s(t)$ that is a step funciton $\theta(t)$ with height $V_0$:

$$
V_s(t) = V_0 \theta(t) 
$$

```python
mpl.rc('font', size=14)
fig = plt.figure(figsize=(7,3), dpi=100)
plt.axis('off')
plt.plot([-1.2,5.2],[0,0], 'k')
plt.plot([0,0],[-0.5,2.5], 'k')
plt.text(5, -0.4,"$t$")
plt.text(-0.5, 2.5,"$V_s$")
plt.text(-0.5, 1.9, "$V_0$")
t = np.linspace(-1,5, 1000)
V0 = (t>0)*2
plt.plot(t,V0, lw=3, c="blue")
show_plot(fig)
```

For $t>0$, $dV_s / dt = 0$, hence we are interested in the solutions of the following equation:

$$
\frac{d^2I}{dt^2} = -  \frac{I}{LC} 
$$

The most general solution of this equation will be a combination of a cosine and a sine:

$$
I(t) = A \cos(\omega_0 t) + B \sin(\omega_0 t)
$$

In contrast to the solution to the step response of the RC circuit and the RL circuit, which decayed exponentially, the step response of an LC circuit will oscillate in time. The frequency of these oscillations of the undamped LC circuit is known as the [natural frequency](https://en.wikipedia.org/wiki/Natural_frequency) and is given by:

$$
\omega_0 = \frac{1}{\sqrt{LC}}
$$

To find the coefficients $A$ and $B$, we need to apply the boundary conditions at $t=0$. Since the voltage source outputs zero volts for all times $t<0$, the current for $t<0$ must be zero (the capacitor is discharged, there is no current flowing in the inductor). When the voltage step occurs at $t=0$, the circuit will want to suddenly allow current to flow. However, since the inductor will not allow sudden changes in current ($V = L dI/dt$), a "back EMF" will develop across the inductor in response to this step to keep the current zero and allow it instead to slowly build up. From this logic, we find that $I(t=0) = 0$, leaving us with only the term sine term:

$$
I(t) = B \sin(\omega_0 t)
$$

To find the coefficient B, we can go back to the original equation before we took the derivative:

$$
V_s  = L \frac{dI}{dt} + \frac{Q}{C} 
$$

Since the charge $Q$ on the capacitor starts at zero at $t=0$ and cannot change quickly due to the current being held back by the inductor, at $t=0$, and using that the derivative of a sine is a cosine, we then get: 

$$
V_0 = L \frac{dI}{dt}|_{t=0} = L (B cos(\omega)t) |_{t=0} = L \omega_0 B
$$

which gives:

$$ 
B = \frac{V_0}{\omega_0 L}
$$

and

$$
I(t) = \frac{V_0}{\omega_0 L} \sin(\omega_0 t)
$$

To calculate the voltage $V_C$ across the capacitor, again, we can either integrate the current to get the charge on the capacitor, or we can subtract the voltage across the inductor, giving:

$$
V_C(t) = V_0 (1 - \cos(\omega_0 t))
$$


```python
fig = plt.figure(figsize=(7,6))
plt.subplot(211)
plt.axis('off')
plt.plot([-0.5,3.2],[0,0], 'k')
plt.plot([0,0],[-2.5,2.5], 'k')
plt.text(3.1, -0.45,"$t$")
plt.text(-0.20, 2.5,"$I$")
plt.text(-0.3, 1.7, r"$\frac{V_0}{\omega_0 L}$")
t = np.linspace(-0.3,3, 1000)
V0 = (t>0)*1.8
plt.plot([0,3.2],[V0,V0], lw=1, ls="--", c='k')
plt.plot([1,1],[0,-1.3], lw=1, ls="--", c='k')
plt.text(0.95, -1.8,r"$t = 2 \pi / \omega_0$", size=12)
plt.plot(t,V0*np.sin(t*2*np.pi), c="red", lw=2)

plt.subplot(212)
plt.axis('off')
plt.plot([-0.5,3.2],[0,0], 'k')
plt.plot([0,0],[-0.4,4], 'k')
plt.text(3.1, -0.45,"$t$")
plt.text(-0.3, 3.8,"$V_C$")
plt.text(-0.3, 1.7, r"$V_0$")
t = np.linspace(-0.3,3, 1000)
V0 = (t>0)*1.8
plt.plot([0,3.2],[V0,V0], lw=1, ls="--", c='k')
plt.plot(t,V0*(1-np.cos(t*2*np.pi)), c="blue", lw=2)

show_plot(fig)
```

In the upper plot, we see that the current begins to build up linearly through the inductor, but then reaches a peak value, before oscillating back down again and becoming negative, and then returning back up to zero after a time $t=2\pi / \omega_0$. The maximum height it reaches is $I_max = V_0 / \omega_0 L$, which is perhaps not coincidentally closely related to the current one would expect from the impedance of the inductor $V_0 / |Z_L|$.

In the lower plot, we see the voltage on the capacitor. As current starts to flow through the inductor, the capacitor charges up and we build up a voltage on the capacitor. Eventually, the current in the inductor inverts, and the capacitor discharges. What is very strking about the plot is that the voltage on the capacitor at its peak is **actually larger** than the voltage supplied by the source: when charging up, the capacitor **overshoots**! As the capacitor charge climbs up, it reaches the value $V_0$ given by the voltage source, but at that point in time, there is still current flowing in the inductor. To slow down the current flowing in the inductor, you need to build up a negative voltage drop across it to slow the current down again, and hence the voltage of the capacitor gets pushed larger than $V_0$ at the peak when the current is zero. 

Important in this story is that neither the inductor nor the capacitor can dissipate energy: the initial energy added to the circuit by the step response of the voltage source will oscillate back and forevever, first being stored in the magentic fields in the inductor, then being transferred to electric fields in the capacitor as the voltage overshoots to slow the current down, then back into magnetic fields in the inductor as the capacitor discharges again, and then back into the capacitor as the negative overshoot of the current comes back up again. Without any dissipation, this oscillation cycles goes on until $t = \infty$.

This oscillation of energy is exactly analagous to the oscillation of the energy of a mass-spring system between the kinetic energy of the mass and the potential energy stored in the spring. This is also the origin of a common name [tank circuit](https://en.wikipedia.org/wiki/LC_circuit) for LC circuits, based on the analogue of water sloshing around in a tank of water on a ship.

### The series RLC circuit

To damp the oscillations we saw in the series RLC circuit, we will add a resistor to the circuit. This could be an actually intentional resistor, but also in practice, unless the wires of your inductor are superconducting, there will always be a (potentially small) series resistance from the wires in the coil of your inductor. This will lead us to the following circuit:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Ground()
    S = elm.SourceV().up().label(r"$V_s(t)$")
    elm.CurrentLabelInline(direction='out', start=False).at(S).label('I')
    elm.Resistor().label("R").right()
    elm.Inductor().label("L").right()
    elm.Dot().label("$V_C$")
    elm.Capacitor().down().label("C")
    elm.Line().left().length(6)
    display_schematic(d)
```

We will use Kirchoff voltage law for the voltage drop around the loop, giving the following equation:

$$
V_s  = IR + L \frac{dI}{dt} + \frac{Q}{C}
$$

We will again take a derivative to get the following second order differential equation for the current:

$$
L \frac{d^2I}{dt^2} + R\frac{dI}{dt} + \frac{1}{C} I  = \frac{dV_s}{dt}
$$

You will likely recognise this as the equation of motion of the [damped harmonic oscillator](https://en.wikipedia.org/wiki/Harmonic_oscillator#Damped_harmonic_oscillator), with the current playing the role of the position $x$, the inductance the role of the mass $m$, $1/C$ playing the role of the spring constant, and the resistance $R$ playing the role of the [viscous damping coefficient](https://en.wikipedia.org/wiki/Viscous_damping#Single-degree-of-freedom_system) $c$. 

We can solve this by ansatz by showing that the solution is given by: 

$$
I(t) = I_0 e^{-\gamma t} \cos(\omega_1 t + \phi)
$$
