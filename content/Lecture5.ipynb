## The equations of motion of RLC circuits

### The series LC circuit 

We will start with a simple circuit consiting an inductor and resistor in series driven by an ideal voltage source:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Ground()
    S = elm.SourceV().up().label(r"$V_s(t)$")
    elm.CurrentLabelInline(direction='out', start=False).at(S).label('I')
    elm.Inductor().label("L").right()
    elm.Dot().label("$V_C$")
    elm.Capacitor().down().label("C")
    elm.Line().left()
    display_schematic(d)
```

Using Kirchoff voltage law for the voltage drop around the loop, we will get the following equation:

$$
V_s  = L \frac{dI}{dt} + \frac{Q}{C}
$$

If we take the derivative of this with repect to time, we can see that this is not a *second order* differential equation for the current $I$:

$$
L \frac{d^2I}{dt^2} + \frac{I}{C} = \frac{dV_s}{dt}
$$

As we did in Lecture 2, we will consider a voltage $V_s(t)$ that is a step funciton $\theta(t)$ with height $V_0$:

$$
V_s(t) = V_0 \theta(t) 
$$

```python
mpl.rc('font', size=14)
fig = plt.figure(figsize=(7,3), dpi=100)
plt.axis('off')
plt.plot([-1.2,5.2],[0,0], 'k')
plt.plot([0,0],[-0.5,2.5], 'k')
plt.text(5, -0.4,"$t$")
plt.text(-0.5, 2.5,"$V_s$")
plt.text(-0.5, 1.9, "$V_0$")
t = np.linspace(-1,5, 1000)
V0 = (t>0)*2
plt.plot(t,V0, lw=3, c="blue")
show_plot(fig)
```

For $t>0$, $dV_s / dt = 0$, hence we are interested in the solutions of the following equation:

$$
\frac{d^2I}{dt^2} = -  \frac{I}{LC} 
$$

The most general solution of this equation will be a combination of a cosine and a sine:

$$
I(t) = A \cos(\omega_0 t) + B \sin(\omega_0 t)
$$

In contrast to the solution to the step response of the RC circuit and the RL circuit, which decayed exponentially, the step response of an LC circuit will oscillate in time. The frequency of these oscillations of the undamped LC circuit is known as the [natural frequency](https://en.wikipedia.org/wiki/Natural_frequency) and is given by:

$$
\omega_0 = \frac{1}{\sqrt{LC}}
$$

To find the coefficients $A$ and $B$, we need to apply the boundary conditions at $t=0$. Since the voltage source outputs zero volts for all times $t<0$, the current for $t<0$ must be zero (the capacitor is discharged, there is no current flowing in the inductor). When the voltage step occurs at $t=0$, the circuit will want to suddenly allow current to flow. However, since the inductor will not allow sudden changes in current ($V = L dI/dt$), a "back-EMF" ([counter electromotive force](https://en.wikipedia.org/wiki/Counter-electromotive_force)) will develop across the inductor in response to keep the current zero at $t=0$ and allow it instead to slowly build up. From this, we can applly a boundary condition that $I(t=0) = 0$, leaving us with only the term sine term:

$$
I(t) = B \sin(\omega_0 t)
$$

To find the coefficient B, we can go back to the original equation before we took the derivative:

$$
V_s  = L \frac{dI}{dt} + \frac{Q}{C} 
$$

Since the charge $Q$ on the capacitor starts at zero at $t=0$ and cannot change quickly due to the current being held back by the inductor, at $t=0$, and using that the derivative of a sine is a cosine, we then get: 

$$
V_0 = L \frac{dI}{dt}|_{t=0} = L (B cos(\omega)t) |_{t=0} = L \omega_0 B
$$

which gives:

$$ 
B = \frac{V_0}{\omega_0 L}
$$

and

$$
I(t) = \frac{V_0}{\omega_0 L} \sin(\omega_0 t)
$$

To calculate the voltage $V_C$ across the capacitor, again, we can either integrate the current to get the charge on the capacitor, or we can subtract the voltage across the inductor, giving:

$$
V_C(t) = V_0 (1 - \cos(\omega_0 t))
$$


```python
fig = plt.figure(figsize=(7,6))
plt.subplot(211)
plt.axis('off')
plt.plot([-0.5,3.2],[0,0], 'k')
plt.plot([0,0],[-2.5,2.5], 'k')
plt.text(3.1, -0.45,"$t$")
plt.text(-0.20, 2.5,"$I$")
plt.text(-0.3, 1.7, r"$\frac{V_0}{\omega_0 L}$")
t = np.linspace(-0.3,3, 1000)
V0 = (t>0)*1.8
plt.plot([0,3.2],[V0,V0], lw=1, ls="--", c='k')
plt.plot([1,1],[0,-1.3], lw=1, ls="--", c='k')
plt.text(0.95, -1.8,r"$t = 2 \pi / \omega_0$", size=12)
plt.plot(t,V0*np.sin(t*2*np.pi), c="red", lw=2)

plt.subplot(212)
plt.axis('off')
plt.plot([-0.5,3.2],[0,0], 'k')
plt.plot([0,0],[-0.4,4], 'k')
plt.text(3.1, -0.45,"$t$")
plt.text(-0.3, 3.8,"$V_C$")
plt.text(-0.3, 1.7, r"$V_0$")
t = np.linspace(-0.3,3, 1000)
V0 = (t>0)*1.8
plt.plot([0,3.2],[V0,V0], lw=1, ls="--", c='k')
plt.plot(t,V0*(1-np.cos(t*2*np.pi)), c="blue", lw=2)

show_plot(fig)
```

In the upper plot, we see that the current begins to build up linearly through the inductor, but then reaches a peak value, before oscillating back down again and becoming negative, and then returning back up to zero after a time $t=2\pi / \omega_0$. The maximum height it reaches is $I_max = V_0 / \omega_0 L$, which is perhaps not coincidentally closely related to the current one would expect from the impedance of the inductor $V_0 / |Z_L|$.

In the lower plot, we see the voltage on the capacitor. As current starts to flow through the inductor, the capacitor charges up and we build up a voltage on the capacitor. Eventually, the current in the inductor inverts, and the capacitor discharges. What is very strking about the plot is that the voltage on the capacitor at its peak is **actually larger** than the voltage supplied by the source: when charging up, the capacitor **overshoots**! As the capacitor charge climbs up, it reaches the value $V_0$ given by the voltage source, but at that point in time, there is still current flowing in the inductor. To slow down the current flowing in the inductor, you need to build up a negative voltage drop across it to slow the current down again, and hence the voltage of the capacitor gets pushed larger than $V_0$ at the peak when the current is zero. 

Important in this story is that neither the inductor nor the capacitor can dissipate energy: the initial energy added to the circuit by the step response of the voltage source will oscillate back and forevever, first being stored in the magentic fields in the inductor, then being transferred to electric fields in the capacitor as the voltage overshoots to slow the current down, then back into magnetic fields in the inductor as the capacitor discharges again, and then back into the capacitor as the negative overshoot of the current comes back up again. Without any dissipation, this oscillation cycles goes on until $t = \infty$.

This oscillation of energy is exactly analagous to the oscillation of the energy of a mass-spring system between the kinetic energy of the mass and the potential energy stored in the spring. This is also the origin of a common name [tank circuit](https://en.wikipedia.org/wiki/LC_circuit) for LC circuits, based on the analogue of water sloshing around in a tank of water on a ship. Note that even though we have only applied an inpulse step voltage, the resulting currents in the circuit oscillate in time: the voltage step is like a hammer that you have hit the circuit with that makes it "ring" like a bell.

### The series RLC circuit

To damp the oscillations we saw in the series RLC circuit, we will add a resistor to the circuit. This could be an actually intentional resistor, but also in practice, unless the wires of your inductor are superconducting, there will always be a (potentially small) series resistance from the wires in the coil of your inductor. This will lead us to the following circuit:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Ground()
    S = elm.SourceV().up().label(r"$V_s(t)$")
    elm.CurrentLabelInline(direction='out', start=False).at(S).label('I')
    elm.Resistor().label("R").right()
    elm.Inductor().label("L").right()
    elm.Dot().label("$V_C$")
    elm.Capacitor().down().label("C")
    elm.Line().left().length(6)
    display_schematic(d)
```

We will use Kirchoff voltage law for the voltage drop around the loop, giving the following equation:

$$
V_s  = IR + L \frac{dI}{dt} + \frac{Q}{C}
$$

We will again take a derivative to get the following second order differential equation for the current:

$$
L \frac{d^2I}{dt^2} + R\frac{dI}{dt} + \frac{1}{C} I  = \frac{dV_s}{dt}
$$

You will likely recognise this as the equation of motion of the [damped harmonic oscillator](https://en.wikipedia.org/wiki/Harmonic_oscillator#Damped_harmonic_oscillator), with the current playing the role of the position $x$, the inductance the role of the mass $m$, $1/C$ playing the role of the spring constant, and the resistance $R$ playing the role of the [viscous damping coefficient](https://en.wikipedia.org/wiki/Viscous_damping#Single-degree-of-freedom_system) $c$. 

We can solve this by ansatz by showing that the solution is given by: 

$$
I(t) = I_0 e^{-\gamma t} \sin(\omega_1 t + \phi)
$$

Plugging this into the equation above, we can see that it is a solution with a damping coefficient $\gamma$ given by:

$$
\gamma = \frac{R}{2L}
$$

The damping-modified natural frequency $\omega_1$ given by:

$$
\omega_1 = \sqrt{\omega_0^2 - \gamma^2} = \sqrt{\frac{1}{LC} - \left(\frac{R}{2L}\right)^2}
$$

If $\omega_0 > \gamma$, then $\omega_1$ is a real-valued number and the solution will indeed be an oscillating sine function like we assumed in our ansatz. 

At $t=0$, since the circuit starts with no charge and no current, like the series LC circuit, the current at $t=0$ must be zero  $I(t=0) = 0$, giving that $\phi = 0$. Again, in the same way as the series LC circuit, balancing the voltages at $t=0$, we will also find that $I_0 = V_0 / \omega_1 L$, since there is also no voltage drop across the resistor. 

We can also find the voltage $V_C(t)$ across the capacitor:

$$
V_C(t) = V_0 - L\frac{dI}{dt} - RI = V_0 \left[ 1 - e^{-\gamma t} \left( \cos(\omega_1 t) + \frac{\gamma}{\omega_1} \sin(\omega_1 t) \right) \right]
$$

Let's take a look at what these look like for $\gamma = 0.1 \omega_1$:

```python
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure(figsize=(7, 6))

# Normalized parameters for plotting (setting omega_0 = 2*pi so T_0 = 1, V_0 = 1.8 for scaling)
omega_0 = 2 * np.pi*4
gamma = omega_0 / 10
omega_1 = np.sqrt(omega_0**2 - gamma**2)
V0_val = 1.8

# Time array
t = np.linspace(-0.3, 3, 1000)
# Step function for voltage source driving
V_step = (t > 0) * V0_val

# 1. Top subplot: Current I(t)
plt.subplot(211)
plt.axis("off")
plt.plot([-0.5, 3.2], [0, 0], "k")
plt.plot([0, 0], [-2.5, 2.5], "k")
plt.text(3.1, -0.45, "$t$")
plt.text(-0.20, 2.5, "$I$")
plt.plot([2*np.pi/omega_1, 2*np.pi/omega_1], [0, -1.3], lw=1, ls="--", c="k")
plt.text(0.3, -1.8, r"$t = 2 \pi / \omega_1$", size=12)
plt.text(0.3, 1.5, r"$e^{-\gamma t}$", size=12)


# Damped current solution for t > 0 (zero for t <= 0)
# Amplitude factor: V_0 / (omega_1 * L), scaled here relative to V0_val
I_t = np.where(
    t > 0,
    (V0_val / (omega_1 * (V0_val / omega_0)))
    * np.exp(-gamma * t)
    * np.sin(omega_1 * t),
    0,
)
plt.plot(t, I_t*2, c="red", lw=2)
t1 = np.linspace(0,3,1000)
plt.plot(t1, (2*(V0_val / (omega_1 * (V0_val / omega_0))) * np.exp(-gamma * t1))*(t1>0), c="k", ls="--", lw=1)


# 2. Bottom subplot: Capacitor Voltage V_C(t)
plt.subplot(212)
plt.axis("off")
plt.plot([-0.5, 3.2], [0, 0], "k")
plt.plot([0, 0], [-0.4, 4], "k")
plt.text(3.1, -0.45, "$t$")
plt.text(-0.3, 3.8, "$V_C$")
plt.text(-0.3, 1.7, "$V_0$")

plt.plot([0, 3.2], [V0_val, V0_val], lw=1, ls="--", c="k")

# Damped capacitor voltage solution for t > 0 (zero for t <= 0)
VC_t = np.where(
    t > 0,
    V0_val
    * (
        1
        - np.exp(-gamma * t)
        * (np.cos(omega_1 * t) + (gamma / omega_1) * np.sin(omega_1 * t))
    ),
    0,
)
plt.plot(t, VC_t, c="blue", lw=2)

show_plot(fig)
```

Looking at the top plot, we see that like in the series LC circuit, the series RLC circuit also responds to the step in voltage by producing an oscillation in current, but now this oscillation does not go on forever, but is instead damped by an exponential envelope $e^{-\gamma t}$. The damping coefficient in the series RLC circuit is proportional to the resistance:

$$
\gamma = \frac{R}{2L}
$$

The larger the series resistance, the faster the damping occurs. A second thing we can notice about the oscillations in the series RLC circuit is that the frequency of oscillation is slighly lower than that of the natural, undamped frequency: $\omega_1  = \sqrt{\omega_0^2 - \gamma^2}$. 

Looking at the lower plot of $V_C(t)$, we see that will also experience damped oscillations, but that at long times, we recover back the step in the voltage up to $V_0$: once the oscillations die out, the capacitor does end up charged to a static voltage. We can see, however, that unlike the RC circuit, where there was just an exponential decay up to this new value, once we add the inductor, the voltage can "overshoot" and oscillate before reaching the new value. This "ringing" in response to steps is an indication that somewhere in your circuit, there is a capacitor and inductor that are resonating with each other. If you have sharp enough pulse, and some stray capacitances and inductances in your circuit, you will very frequently see this type of "overshoot" and ringing, an it is a common engineering challenge to get rid of them in high frequency electronics. 

In the plot above, we have chosen a ratio of $\gamma / \omega_0 = 0.1$: the smaller this ratio, the more oscillations you will see before they decay. This is commonly quantified in something called the [quality factor](https://en.wikipedia.org/wiki/Q_factor) defined as:

$$
Q = \frac{\omega_0}{2 \gamma} 
$$

Note that the factor of two here comes from the fact that the quality factor is defined in terms of the energy decay rate: since the energy in the inductor is proportional to $I^2$, the energy decays at a rate $2 \gamma$ based on our convention above. 

For the series RLC, the Q-factor is given by:

$$
Q_{\rm ser} = \frac{1}{R} \sqrt{\frac{L}{C}} = \frac{Z_C}{R}
$$

where $Z_C = \omega_0 L = 1 / (\omega_0 C)$ is a number known as the *characteristic impedance* of the resonator. As we discussed above, for the series RLC circuit, a lower resistance will give a higher Q. The reason for this is that when the circuit is resonating, the resonant current is always forced to flow through the resistor, and so the higher the resistance, the more dissipation, since for a resistor, $P = I^2 R$. The plots above are made with $Q = 1 / 0.1 / 2 = 5$. In the plot you can also "see" about 5 oscillations, which is a "finger in the air" way to estimate the Q-factor from the ringdown plot. For quantitative numbers, you can better fit, or estimate $\gamma$ from the 1/e time of the decay envelope. 

The story above so far with the ringing oscillations holds only if the damping is sufficiently small ($\gamma \ll \omega_0$). If that is not the case, you end up with different solutions: if $\gamma > \omega_0$, then $\omega_1 = \sqrt{\omega_0^2 - \gamma^2}$ becomes imaginary: the sine become a hyperbolic sine and there are no oscillations any more:

```python
import matplotlib.pyplot as plt
import numpy as np

# Parameters
omega_0 = 2 * np.pi
t = np.linspace(-0.3, 3, 1000)

gamma_values = [0.1 * omega_0, 1.0 * omega_0, 10.0 * omega_0]
titles = [
    r"Underdamped ($\gamma = 0.1\omega_0$)",
    r"Critically Damped ($\gamma = \omega_0$)",
    r"Overdamped ($\gamma = 10\omega_0$)",
]

fig, axes = plt.subplots(1, 3, figsize=(10, 3))

for ax, gamma, title in zip(axes, gamma_values, titles):
  plt.sca(ax)
  ax.axis("off")

  # Calculate raw current response
  if gamma < omega_0:
    omega_1 = np.sqrt(omega_0**2 - gamma**2)
    I_t = np.where(t > 0, np.exp(-gamma * t) * np.sin(omega_1 * t), 0)
  elif np.isclose(gamma, omega_0):
    I_t = np.where(t > 0, t * np.exp(-gamma * t), 0)
  else:
    s1 = -gamma + np.sqrt(gamma**2 - omega_0**2)
    s2 = -gamma - np.sqrt(gamma**2 - omega_0**2)
    I_t = np.where(t > 0, np.exp(s1 * t) - np.exp(s2 * t), 0)

  # Normalize each curve independently so it fills the same visual height
  max_val = np.max(np.abs(I_t))
  if max_val > 0:
    I_t_normalized = I_t / max_val
  else:
    I_t_normalized = I_t

  # Plot axes centered symmetrically (y from -1.5 to 1.5)
  ax.plot([-0.5, 3.2], [0, 0], "k")
  ax.plot([0, 0], [-1.5, 1.5], "k")
  ax.set_ylim(-1.5, 1.5)

  ax.text(3.1, -0.15, "$t$")
  ax.text(-0.20, 1.4, "$I$")
  ax.set_title(title, fontsize=12, pad=15)

  ax.plot(t, I_t_normalized, c="red", lw=2)

plt.tight_layout()
show_plot(fig)
```

In the underdamped limit, the current oscillates as we discussed above. For critical damping, corresponding to $\gamma = \omega_0$, there is exactly one bump followed by a decay. In the overdamped limit $\gamma >> \omega_0$, there is first a fast rise in the current with a time constant $L/R$ associated with the energisation of the inductor, followed by a long decay with a time constant $RC$ associated with the charging of the capacitor. The overdamped regime looks a lot like the RC circuit we saw in lecture 3, except that in the theoretical ideal RC circuit model, the jump in the current at $t=0$ is really an instantaneous step function, while here with the inductor is it now smoothed out by the $L/R$ time constant.

### The parallel RLC

We can also combine the resistor, inductor, and capacitor in parallel instead of in series. Specifically, we will consider a parallel RLC circuit driven by an ideal current source that provides a step in the current from 0 to $I_0$ at $t = 0$:


```python
with schemdraw.Drawing(show=False) as d:
    elm.Ground()
    S = elm.SourceI().up().label(r"$I_s(t)$")
    elm.Line().right()
    d.push()
    elm.Resistor().label("R").down()
    d.pop()
    elm.Line().right().length(2)
    d.push()
    elm.Inductor().label("L").down()
    d.pop()
    elm.Line().right().length(2)
    elm.Dot().label("$V(t)$")
    elm.Capacitor().label("C").down()
    d.pop()
    elm.Line().left().length(7)
    display_schematic(d)
```

And now we will ask the question: what is the solution for V(t)? 

We will leave it as a homework exercise to show that the solution in the underdamped regime is given by:

$$
V(t) = \frac{I_0}{\omega_1 C} e^{-\gamma t} \sin(\omega_1 t)
$$

Note that the voltage now oscillates to zero and there is no step: at long times, the inductor short circuits the resistor and capacitor giving no DC output voltage. The damping-modified oscillation frequency is once again given by: 

$$
\omega_1 = \sqrt{\omega_0^2 - \gamma^2}
$$

For the parallel RLC, though, we will have a damping rate:

$$
\gamma_{\rm par} = \frac{1}{2RC}
$$

If you look back, you will remember that the resitance $R$ was in the numerator for the series RLC circuit: bigger resistance gave more damping. Here, the resistance is in the denominator: bigger resistance gives *less* damping! While that may seem at first counter intuitive, if you think about how currents will flow in the parallel RLC, then you can understand what is going on: if the resistance is smaller in the parallel RLC, then *more* current will flow through the resistor, given *more* damping. If the resitance is infinite, then *no* current flows through the resistor: all currents flow only the through the non-dissipative inductor and capacitor, and there is no damping at all. 

This also comes back if you look at the formula for the Q-factor of the parallel RLC circuit: 

$$
Q_{\rm par} = \frac{R}{Z_C} = \frac{1}{Q_{\rm ser}}
$$

Again, you see here the opposite behaviour: the parallel RLC circuit has a high quality factor for a large resistance, while exactly the opposite is true for the series RLC circuit. 

## The impedance and frequency response RLC circuits

We will now switch gears and take a look at the response of (R)LC circuits to an oscillating driving voltage or current. We will do this, of course, using the framework of complex impedance. 

### Coherent cancellation of currents in parallel and series LC circuits

We will first consider a series combination of an inductor and a capacitor and determine the effective combined impedance of the circuit:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Inductor().label("L").right().at((0,0))
    elm.Capacitor().right().label("C")
    elm.ResistorIEC().label(r"$Z_{\rm eq}$").right().at((9,0))
    elm.Arrow().length(1).at((7,0))
    display_schematic(d)
    
```

Impedances sum in series and so the equivalent impedance of this series combination is:

$$
Z_{\rm eq,ser} = j \omega L + \frac{1}{j \omega C} 
= j \omega L \left(1 - \frac{1}{\omega^2 L C}\right)
= j \omega L \left(1 - \frac{\omega_0^2}{\omega^2}\right)
$$

where $\omega_0$ is the natural frequency again. Since we have only non-dissipative elements, the impedance is purely imaginary valued. It is useful to plot ${\rm Im}(Z_{\rm eq})$ to see what it looks like:

```python
fig = plt.figure(figsize=(6,4))
plt.axis("off")
xlim = (-0.5,3)
ylim = (-1,1)
plt.plot(xlim, [0, 0], "k") #x axis
plt.plot([0, 0], ylim, "k") #y axis
plt.xlim(xlim)
plt.ylim(ylim)
plt.text(0, ylim[1]*1.1, r"${\rm Im}(Z_{\rm eq, ser})$", ha='center')
plt.text(xlim[1]*1.05, 0, r"$\omega / \omega_0$", va='center')

plt.plot((1,1),(0.2,-0.2), "k:")

x = np.linspace(0.1,xlim[1],100)
plt.plot(x, (1-1/x**2), c = "blue")
show_plot(fig)
```

At low frequencies, the impedance is dominated by the $1/\omega$ dependence of the capacitor, while at high frequencies, the lineary scaling of the impedance with frequency from the inductor takes over. 

A super imporant point of this plot is the difference in **sign** of the impedance of the inductor and capacitor: the capacitor contributes a negative imaginary impedance, while the inductor contributes a positive imaginary impedance. 

This difference in sigh leads to a very special point at $\omega = \omega_0$: at the natural resonance frequency, the two impedences **cancel**! Even though both the capacitor and inductor on their own have non-zero impedance at $\omega_0$, when you put them in series, these two **exactly cancel** and the series combination goes to zero!!! The circuit looks like a short circuit at exactly one frequency where the magnitude of the impedance vanishes:

```python
fig = plt.figure(figsize=(6,4))
plt.axis("off")
xlim = (-0.5,3)
ylim = (-0.2,1)
plt.plot(xlim, [0, 0], "k") #x axis
plt.plot([0, 0], ylim, "k") #y axis
plt.xlim(xlim)
plt.ylim(ylim)
plt.text(0, ylim[1]*1.1, r"$|Z_{\rm eq,ser}|$", ha='center')
plt.text(xlim[1]*1.05, 0, r"$\omega / \omega_0$", va='center')

plt.plot((1,1),(0,-0.1), "k:")
plt.text(1,-0.2, r"$\omega = \omega_0$", ha="center")

x = np.linspace(0.1,xlim[1],1000)
plt.plot(x, np.abs((1-1/x**2)), c = "magenta")
show_plot(fig)
```

The series LC circuit is has a very strange behaviour that it will resist the flow of current at all frequencies, except exactly at the resonance frequency, where it will allow infinite current to flow for any applied voltage.

We can also look at what happens when we put the inductor and capacitor in parallel:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Line().length(1).right().at((0,0))
    d.push()
    elm.Line().length(0.5).up()
    elm.Inductor().label("L").right()
    elm.Line().length(0.5).down()
    elm.Line().length(1).right()
    d.pop()
    elm.Line().length(0.5).down()
    elm.Capacitor().right().label("C", loc="bottom")
    elm.Line().length(0.5).up()
    
    elm.Arrow().right().length(1).at((6,0))

    elm.ResistorIEC().label(r"$Z_{\rm eq}$").right().at((8,0))
    
    display_schematic(d)  
```

Parallel impedances add in inverse, giving:

$$
\frac{1}{Z_{\rm eq, par}} = j \omega C + \frac{1}{j \omega L} = \frac{1- \omega^2 LC}{j \omega L}
$$

Inverting this, we get:

$$
Z_{\rm eq, par} = \frac{j \omega L}{1- \omega^2 LC}
$$

In constrast the vanishing impedance of the series LC circuit at resonance, the impedance of the parallel LC circuit diverges at resonance:

```python
fig = plt.figure(figsize=(12,4))

plt.subplot(121)
plt.axis("off")
xlim = (-0.5,3)
ylim = (-10,10)
plt.plot(xlim, [0, 0], "k") #x axis
plt.plot([0, 0], ylim, "k") #y axis
plt.xlim(xlim)
plt.ylim(ylim)
plt.text(0, ylim[1]*1.1, r"${\rm Im}(Z_{\rm eq, par})$", ha='center')
plt.text(xlim[1]*1.05, 0, r"$\omega / \omega_0$", va='center')
plt.plot((1,1),(-10,10), "k:")

x1 = np.linspace(0.1,0.999,100)
x2 = np.linspace(1.001,xlim[1],100)
plt.plot(x1, 1/(1-1/x1**2), c = "blue")
plt.plot(x2, 1/(1-1/x2**2), c = "blue")

plt.subplot(122)
plt.axis("off")
xlim = (-0.5,3)
ylim = (-1,10)
plt.plot(xlim, [0, 0], "k") #x axis
plt.plot([0, 0], ylim, "k") #y axis
plt.xlim(xlim)
plt.ylim(ylim)
plt.text(0, ylim[1]*1.1, r"$|Z_{\rm eq,par}|$", ha='center')
plt.text(xlim[1]*1.05, 0, r"$\omega / \omega_0$", va='center')
plt.plot((1,1),(-1,10), "k:")

x1 = np.linspace(0.1,0.999,100)
x2 = np.linspace(1.001,xlim[1],100)
plt.plot(x1, -1/(1-1/x1**2), c = "magenta")
plt.plot(x2, 1/(1-1/x2**2), c = "magenta")

show_plot(fig)
```

Away from resonance, the series LC allows current to flow, but at resonance, the currents flowing through the capacitor and inductor exactly cancel each other and all current is blocked.
