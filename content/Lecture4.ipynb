## AC Voltages, Currents and Power

In the previous lecture, we started explicitly looking at time dependent voltages and currents in circuits by deriving the differential equations that describe currents and voltages in circuits, and solving the response of RC and RL circuits to a step function input voltage. 

In this lecture, we will explore the response of circuits to an oscillating voltage:

$$
V_s(t) = V_0 \cos(\omega t)
$$

or current:

$$
I_s(t) = I_0 \cos(\omega t)
$$

where $\omega = 2 \pi f$ is the angular frequency of the oscillating signal (in rad/sec) and $f$ is the (rotational) frequency of the oscillating signal in Hertz. 

### Basic concepts: Amplitude, PTP, and RMS values

Before we start, it will be useful to go over the ways we quantify the size of AC signals. For this, there are three common conventions (and a forth that we will cover later):

```python
plt.rcParams["text.usetex"] = True
plt.rcParams["font.family"] = "serif"

figsize = np.array((7,2))*0.8
fig = plt.figure(figsize=figsize, dpi=100)
plt.axis('off')
t = np.linspace(-0,3, 1000)
plt.xlim(-0.8,t[-1]+3)
plt.plot([t[0],t[-1]],[0,0], 'k')
plt.plot([t[0]-0.25,t[-1]+0.25],[0,0], 'k:')
plt.text(t[-1]+0.35, -0.08,"$t$")

plt.plot([t[0]-0.25,t[-1]+0.25],[1,1], 'k--', lw=0.5)
plt.plot([t[0]-0.25,t[-1]+0.25],[-1,-1], 'k--', lw=0.5)

plt.arrow(-0.5,0,0,1, head_width = 0.05, length_includes_head = True, color='k')
plt.text(-0.8, 0.45, "$V_0$")

plt.arrow(3.7,0,0,1, head_width = 0.05, length_includes_head = True, color='k')
plt.arrow(3.7,0,0,-1, head_width = 0.05, length_includes_head = True, color='k')
plt.text(3.9, -0.08, "$V_{PTP}$")

plt.text(5.4,0.8, r"$V(t) = V_0 \cos(\omega t)$", ha='center')
plt.text(5.4,0.3, r"$V_{\rm amp} = V_0$", ha='center')
plt.text(5.4,-0.25, r"$V_{\rm PTP} = 2V_0$", ha='center')
plt.text(5.4,0.-0.8, r"$V_{\rm RMS} = \frac{1}{\sqrt{2}}V_0$", ha='center')


V0 = -np.cos(t*2*np.pi)
plt.plot(t,V0, lw=1, c="k")
show_plot(fig, scale=2)

plt.rcParams["text.usetex"] = False
plt.rcParams["font.family"] = "sans-serif"
```

For a signal $V_0\cos(\omega t)$, there are three ways we can discuss how big this signal is:

* **Amplitude**: The (peak) amplitude of the signal is $V_{\rm amp} = V_0$, given by hight of the peaks relative to the average value
* **Peak-to-peak (PTP) amplitude:**: The peak to peak amplitude is $V_{\rm PTP} = 2V_0$, given by the hight between the peak and dip voltage
* **Root-mean-square (RMS) amplitude:** The RMS amplitude is $V_{\rm RMS} = \frac{1}{\sqrt{2}}V_0$

(Regular / Peak) Amplitude is the most convenient and obvious way to think about amplitudes when you are working with the mathematical formulas of the signal. 

Peak-to-peak amplitudes are very convenient for working with in the lab, as you can read them off easily from the oscilloscope using cursor modes. Also, oscilloscopes often include the ability to work in AC coupling mode, which automatically removes the DC offset, making it easier to read off the peak-to-peak amplitude with cursors. 

RMS (Root mean square) amplitudes were discussed already in the first lecture, and is motivated by the definition of the power of AC signals: if the above AC voltage is fed into a resistor, that resistor will dissipate a power $P = \frac{1}{2} V_0^2/R = V_{\rm RMS}^2 / R$. When working with RMS voltages, you do not have to remember to include the factor of 1/2 when calculating the power. RMS measures of AC signals is quite common in household voltages, where powers are important for counting how much money you need to pay to the energy supplier, and also in high frequency (RF and microwave) signals, where voltages themselves are hard to measure but powers are easier to detect.  

### AC voltages, AC currents, and average power in inductors and capacitors

Using the results from lecture 2, we can already calculate the relationship between the AC currents flowing in inductors and capacitors, and the voltage that then drops across them. 

We can first consider a capacitor. If we have a capacitor to which we apply a voltage $V(t) = V_0 \cos(\omega t)$, we will have current flowing that is given by:

$$
I = C \frac{dV}{dt} = - \left( C V_0 \omega \right) \sin(\omega t)
$$

We can recognize some things here already: if $\omega \rightarrow 0$, then the current through the capacitor also goes to zero, reflecting the fact that capacitors will not carry current if you expose them to a DC voltage. 

A second intersting thing you can see is that if the voltage across the capacitor is a cosine, then the current across the capacitor is a sine: there is exactly a 90 degree phase lag between the two. A capacitor will only allow current to flow when the voltage across it is changing. 

We can also consider an inductor. If we have an inductor for which a current $I = I_0 \cos(\omega t)$ is flowing, then we will get a voltage across the inductor given by: 

$$
V = L \frac{dI}{dt} = - \left( I_0 L \omega \right) \sin(\omega t)
$$

Again, familiar is that for a DC current, the voltage across the inductor goes to zero. And we also see that the current and voltage 90 degrees out of phase. 

These are both plotted here:

```python
mpl.rc('font', size=14)
fig = plt.figure(figsize=(14,3), dpi=100)

plt.subplot(121)
plt.axis('off')
plt.plot([-1.2,5.2],[0,0], 'k')
plt.plot([0,0],[-1.5,1.5], 'k')
plt.text(5, -0.4,"$t$")
t = np.linspace(-1,5, 1000)
w = 2.5
V0 = np.cos(w*t)
plt.plot(t,V0, lw=3, c="blue")
V0 = -np.sin(w*t)
plt.plot(t,V0, lw=3, c="red")
plt.text(2.6,1.1, "$V(t)$", c="blue")
plt.text(1.3,1.1, "$I(t)$", c="red")
plt.text(2, -2.2, 'Current "leads" (hits peak first)', ha='center', fontsize=20)

plt.title("Capacitor")

plt.subplot(122)
plt.axis('off')
plt.plot([-1.2,5.2],[0,0], 'k')
plt.plot([0,0],[-1.5,1.5], 'k')
plt.text(5, -0.4,"$t$")
t = np.linspace(-1,5, 1000)
w = 2.5
V0 = -np.sin(w*t)
plt.plot(t,V0, lw=3, c="blue")
V0 = np.cos(w*t)
plt.plot(t,V0, lw=3, c="red")
plt.title("Inductor")
plt.text(1.3,1.1, "$V(t)$", c="blue")
plt.text(2.6,1.1, "$I(t)$", c="red")
plt.text(2, -2.2, 'Current "lags" (hits peak later)', ha='center', fontsize=20)

show_plot(fig, scale=0.8)
```

In one case, the current curve "leads", reaching a peak earlier in time than the voltage, and in the other case the current "lags": you have to wait longer for the current to reach the peak after the voltage has already peaked.

What is interesting is that in both cases is that because the current and voltage are shifted from each other by 90 degrees in phase, the time average power over a cycle in both cases is zero:

$$
P_{avg} \propto \int_0^{T = 1/\omega} \cos(\omega t) \sin(\omega t) dt = 0
$$

This is a reflection that both inductors and capacitors only *store* energy and do not dissipate it: on half of the AC cycle, they store energy, and then on the other half of the cycle, they deliver it back.

## Steady-state solutions of the differential equations of RC and RL circuits

In Lecture 3, we used Kirchoff's laws to derive differential equations that describe the voltages and currents in RC and RL circuits for arbitrary driving voltages $V_0(t)$, and solved them for the cases of driving voltages that are in the form of a step function (in other words, we solved for the [step response](https://en.wikipedia.org/wiki/Step_response) of the circuit). 

In the following sections, we will solve these differential equations for an sinusoidally oscillating (AC) driving voltage, which will physically represent the [steady state](https://en.wikipedia.org/wiki/Steady_state_(electronics)) response of the circuit. We will first go through the (relatively simple) mathematical solution of the differential equations for oscillation driving currents or voltages, and then revisit afterwards the physical meaning of the results and the concept of "steady state".  

### Solution of a RC circuit driven by an oscillating voltage

We will consider the RC circuit of lecture 3 but now driven by an AC voltage:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Ground()
    elm.Dot()
    S = elm.SourceV().up().label(r"$V_s(t) = V_0 \cos(\omega t)$")
    elm.CurrentLabelInline(direction='out', start=False).at(S).label('I')
    elm.Dot().label("$V_1$")
    elm.Resistor().label("R").right()
    elm.Dot().label("$V_2$")
    elm.Capacitor().down().label("C")
    elm.Line().left()
    display_schematic(d)
```

In lecture 3, we derived that the differential [equation of motion](https://en.wikipedia.org/wiki/Equations_of_motion) describing this circuit is given by: 

$$
R \frac{dI}{dt} + \frac{I}{C} = \frac{dV_s}{dt}
$$

Using $dV_s/dt =  \omega V_0 \sin(\omega t)$, this becomes:

$$
R \frac{dI}{dt} + \frac{I}{C} = \omega V_0 \sin(\omega t)
$$

Like a good physicist, we will solve this by [ansatz](https://en.wikipedia.org/wiki/Ansatz): we will make the ansatz that the solution will be of the form $I(t) = I_0 \cos(\omega t - \phi)$ and see if we can find values of $I_0$ and $\phi$ that fulfil the equation:

$$
\omega R I_0 \sin(\omega t - \phi) + \frac{1}{C} I_0 \cos(\omega t - \phi) = \omega V_0 \sin(\omega t)
$$

In general, the amplitude and phase of the current will depend on the driving frequency. Both $I_0$ and $\phi$ will depend on $\omega$, and our goal will be to find $I_0(\omega)$ and $\phi(\omega)$.

*(Note: we have chosen the negative sign in from of $\phi$ in our ansatz with a bit of foresight so that we end up with a positive value for I_O, this comes from the fact that we can expect the current of the capacitor to "lag" the voltage across it.)*

How do solve this? We will use the familiar [trigonometric identities](https://en.wikipedia.org/wiki/List_of_trigonometric_identities#Angle_sum_and_difference_identities):

$$
\sin (A+B) = \sin(A)\sin(B) + \cos(A) \cos(B)\\
\cos (A+B) = \cos(A)\cos(B) - \sin(A) \sin(B)
$$

This allows us to rewrite the above as:

$$
\cos(\omega t) \left( \omega R I_0 \sin\phi + \frac{I_0}{C} \cos\phi \right) + \sin(\omega t) \left( -\omega R I_0 \cos\phi + \frac{I_0}{C} \sin\phi \right) &= \omega V_0 \sin(\omega t)
$$

For this equation to hold at all times, the coefficient of the oscillating cosine, and the oscillating sine, on both sides of the equation must be equal.  Considering the cosine term:

$$
\omega R I_0 \sin\phi + \frac{I_0}{C} \cos\phi = 0
$$

we can solve for $\phi(\omega)$:

$$
\tan\phi(\omega) = \frac{1}{\omega R C}
$$


Considering the sine term, we get:

$$
-\omega R I_0 \cos\phi + \frac{I_0}{C} \sin\phi = \omega V_0
$$

We can solve by subtituting in the expressions:

$$
\sin\phi = \frac{1}{\sqrt{1 + \omega^2 R^2 C^2}}, \quad \cos\phi = \frac{\omega R C}{\sqrt{1 + \omega^2 R^2 C^2}}
$$

which in the end will give us the frequency dependent amplitude of the current:

$$
I_0(\omega) = \frac{\omega V_0 C}{\sqrt{1 + \omega^2 R^2 C^2}}
$$

Now that we have done some math, it is a good point to step back, plot what we have derived, and see if we can understand our results based on our physical intution. Here is what these equations predict for the magintude and the phase of the current:

```python
import matplotlib.pyplot as plt
import numpy as np

# Circuit parameters
R = 1000.0  # Resistance in ohms (1 kOhm)
C = 1e-6  # Capacitance in farads (1 uF)
V0 = 1.0  # Driving voltage amplitude

# Characteristic corner frequency omega_0 = 1 / (R * C)
omega_0 = 1.0 / (R * C)

# Frequency range normalized: omega / omega_0 from 10^-3 to 10^3
norm_omega = np.logspace(-3, 3, 1000)
omega = norm_omega * omega_0

# Calculate amplitude (I_m) and phase (phi in degrees) from our derived formulas
I_magnitude = (omega * V0 * C) / np.sqrt(1 + (omega * R * C) ** 2)
phi_deg = np.degrees(np.arctan(1 / (omega * R * C)))

# Create side-by-side subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Magnitude (Amplitude) Bode Plot
ax1.semilogx(norm_omega, I_magnitude * 1e3, color="b", linewidth=2)
ax1.set_title(r"Amplitude of current")
ax1.set_xlabel(r"Normalized Frequency $\omega / \omega_c$")
ax1.set_ylabel(r"Normalised Amplitude $I_0(\omega) / I(\infty)$")
ax1.grid(True, which="both", linestyle="--", alpha=0.7)

# Phase Bode Plot
ax2.semilogx(norm_omega, phi_deg, color="r", linewidth=2)
ax2.set_title(r"Phase lag of current")
ax2.set_xlabel(r"Normalized Frequency $\omega / \omega_c$")
ax2.set_ylabel(r"Phase $\phi$ (degrees)")
ax2.set_ylim(-5, 95)
ax2.set_yticks(np.arange(0, 91, 15))  # Sets y-ticks in increments of 15 degrees from 0 to 90
ax2.grid(True, which="both", linestyle="--", alpha=0.7)

plt.tight_layout()
show_plot(fig)
```

Let's look first at the amplitude of the current: if we go to zero frequency, the current goes to zero. This makes sense, since at low frequencies the capacitor will "block" the current in the circuit. 

At high frequencies, the capacitor becomes a short circuit and then the current becomes frequency independent and determined only by the value of the resistor. We can check this with our formula:

$$
I(\omega \rightarrow \infty) = \frac{\omega V_0 C}{\omega RC} = \frac{V_0}{R}
$$

We can also understand the phase lag plot with the same intuition: at low frequencies, the capacitor is the one that is limiting the current through the circuit, and so the phase lag is 90 degrees, as we would expect for a capacitor. At high frequencies, the capacitor becomes no longer relevant for determining what the current in the circuit, and the phase lag matches the expected in-phase current response of a resistor. 

At frequency we call the "corner" frequency $\omega_c$:

$$
\omega_c = \frac{1}{RC}
$$

the phase lag is exactly 45 degrees, and the current amplitude is $\frac{1}{\sqrt{2}} \approx 0.71$ of its high frequency limited value.

Finally, we can calculate the voltage $V_2$ across the capacitor at the "output" of the circuit. For this, we can either use $V_2(t) = Q(t)/C = 1/C \int I(t) dt$, or we can use $V_2(t) = V_s(t) - I(t)R$. Writing $V_2(t) = V_C(\omega) \cos(\omega_t - \phi_C(\omega))$, we will get:

$$
V_C(\omega) = \frac{V_0}{\sqrt{1 + \omega^2 R^2 C^2}}
$$

$$
\tan \phi(\omega) = \omega R C
$$

```python
import matplotlib.pyplot as plt
import numpy as np

# Circuit parameters
R = 1000.0  # Resistance in ohms (1 kOhm)
C = 1e-6  # Capacitance in farads (1 uF)
V0 = 1.0  # Driving voltage amplitude

# Characteristic corner frequency omega_0 = 1 / (R * C)
omega_0 = 1.0 / (R * C)

# Frequency range normalized: omega / omega_0 from 10^-3 to 10^3
norm_omega = np.logspace(-3, 3, 1000)
omega = norm_omega * omega_0

# Calculate normalized voltage ratio (V_C / V_0) and phase (phi_C in degrees)
V_ratio = 1.0 / np.sqrt(1 + norm_omega ** 2)
phi_C_deg = np.degrees(np.arctan(norm_omega))

# Create side-by-side subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Magnitude Ratio Bode Plot for V_2 / V_0
ax1.semilogx(norm_omega, V_ratio, color="b", linewidth=2)
ax1.set_title(r"Amplitude of capacitor voltage")
ax1.set_xlabel(r"Normalized Frequency $\omega / \omega_c$")
ax1.set_ylabel(r"Voltage Ratio $V_C / V_0$")
ax1.grid(True, which="both", linestyle="--", alpha=0.7)

# Phase Bode Plot for V_2
ax2.semilogx(norm_omega, phi_C_deg, color="r", linewidth=2)
ax2.set_title(r"Phase of capacitor voltage")
ax2.set_xlabel(r"Normalized Frequency $\omega / \omega_c$")
ax2.set_ylabel(r"Phase $\phi_C$ (degrees)")
ax2.set_ylim(-5, 95)
ax2.set_yticks(np.arange(0, 91, 15))  # Increments of 15 degrees from 0 to 90
ax2.grid(True, which="both", linestyle="--", alpha=0.7)

plt.tight_layout()
show_plot(fig)
```

This we can also understand intuitively: at low frequencies, no current flows in the capacitor, it acts like an open circuit, and the voltage from the source is replicated directly at the output. At higher frequencies, the capacitor starts to become a short circuit: you can think of it like a voltage divider that is frequency dependent as the capacitor allows current to flow at higher frequencies, with the division ratio dropping. You will also develop a 90 degree phase shift of the voltage across the capacitor as it starts to allow current to flow. 

### The RC circuit as a filter, The Bode Plot, and the Decibel scale

The RC circuit above we can think of as a [filter](https://en.wikipedia.org/wiki/Electronic_filter) than transforms the "input" voltage of the source into an "output" voltage across the capacitor. In the case of the RC filter, for frequency components above the corner frequency (also sometimes called the cutoff frequency), the signal is attenuated and those frequency components are reduced in the output. 

When thinking about circuits as a filter, one will typically choose to plot the amplitude and phase of the response in a format known as a [Bode plot](https://en.wikipedia.org/wiki/Bode_plot). Bode plots look very similar to the plots we made above except that the y-axis of the plot showing the amplitude response is chosen on a [Decibel](https://en.wikipedia.org/wiki/Decibel) scale. 

For a voltage signal $V$, the decibel scale is defined as:

$$
{\rm dB} = 20 \log_{10} \left( \frac{V}{V_{\rm ref}} \right)
$$

Some useful things about decibels to have handy in your head:

* A voltage ratio of 10 is 20 dB
* A voltage ratio of 1/10 is -20 dB
* 6 dB is a voltage ratio of nearly a voltage ratio of 2 (full value 1.995)

The [wikipedia page](https://en.wikipedia.org/wiki/Decibel#Definition) has a decibel table which can be handy for converting votlage ("amplitude") ratios (and "power" rations) into decibels. 

Note that in order to define a decibel number, you must always either implicitly or explicitly have chosen a definition for the reference value. For the case of our filter above, we have a very clear reference signal $V_{\rm ref} = V_0$ from the ampltude of the input signal $V_s(t) = V_0 \cos(\omega t)$. The decibel measure of the transfer function of the RC filter is then:

$$
{\rm dB} 
= 20 \log_{10} \left( \frac{V_C}{V_0} \right)
= 20 \log_{10} \left( \frac{1}{\sqrt{1 + \omega^2 R^2 C^2}} \right)
$$

Let's plot it:

```python
import matplotlib.pyplot as plt
import numpy as np

# Circuit parameters
R = 1000.0  # Resistance in ohms (1 kOhm)
C = 1e-6  # Capacitance in farads (1 uF)
V0 = 1.0  # Driving voltage amplitude

# Characteristic corner frequency omega_0 = 1 / (R * C)
omega_0 = 1.0 / (R * C)

# Frequency range normalized: omega / omega_0 from 10^-3 to 10^3
norm_omega = np.logspace(-3, 3, 1000)
omega = norm_omega * omega_0

# Calculate normalized voltage ratio (V_C / V_0) and phase (phi_C in degrees)
V_ratio = 1.0 / np.sqrt(1 + norm_omega ** 2)
dB = 20*np.log10(V_ratio)
phi_C_deg = np.degrees(np.arctan(norm_omega))

# Create side-by-side subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Magnitude Ratio Bode Plot for V_2 / V_0
ax1.plot([1e-3,1],[-3,-3], ls=":", c='k')
ax1.plot([1,1],[-60,-3], ls=":", c='k')
ax1.plot([10,100],[-20,-20], ls="--", c='k')
ax1.plot([100,100],[-20,-40], ls="--", c='k')
ax1.set_title(r"Bode Amplitude Plot of RC filter")
ax1.semilogx(norm_omega, dB, color="b", linewidth=2)
ax1.set_xlabel(r"Normalized Frequency $\omega / \omega_c$")
ax1.set_ylabel(r"Relative Amplitude (dB)")
ax1.grid(True, which="both", linestyle="--", alpha=0.7)
ax1.text(1e-3,-8, "3 dB Corner Frequency",weight="bold")
ax1.text(10,-17, "Slope:\n20 dB/ decade",weight="bold")

# Phase Bode Plot for V_2
ax2.set_title(r"Bode Phase Plot of RC filter")
ax2.semilogx(norm_omega, phi_C_deg, color="r", linewidth=2)
ax2.set_xlabel(r"Normalized Frequency $\omega / \omega_c$")
ax2.set_ylabel(r"Relative Phase (degrees)")
ax2.set_ylim(-5, 95)
ax2.set_yticks(np.arange(0, 91, 15))  # Increments of 15 degrees from 0 to 90
ax2.grid(True, which="both", linestyle="--", alpha=0.7)

plt.tight_layout()
show_plot(fig)
```

On the decibel scale, the RC filter will start "roll off" at a slope of 20 dB per decade (about 6 dB per octave) for frequencies about the "3 dB corner frequency" given by $\omega_c$. At the corner frequency, the ratio of the voltages is $\frac{1}{\sqrt{2}}$, which is -3.01 dB (close to 3 dB). The [power ratio](https://en.wikipedia.org/wiki/Decibel#Root-power_(field)_quantities), which tells you how the ratio of the powers that would be dissipated if this voltage was fed into a resistor, and which is therefore proportional to $V^2$, is $\frac{1}{2}$ at the corner frequency.

As a final note on the decibel scale, one can also make a "unit" out of decibles by choosing a particular fixed reference. The [convention](https://en.wikipedia.org/wiki/Decibel#List_of_suffixes) is then to add a letter after the dB letters. Some examples:

#### dBV: Decibels relative to 1 Volt RMS.

For a signal $V_0 \cos(\omega t)$, then:

$$
{\rm dBV} = 20 \log_{10} \left( \frac{\tfrac{1}{\sqrt{2}}V_0}{1\ {\rm V}} \right)
$$

The factor of $1/\sqrt{2}$ comes from the fact that dBV is defined in terms of RMS amplitude. 

#### dBm: Decibels relative to 1 milliwatt (into 50 ohms)

The [dBm](https://en.wikipedia.org/wiki/DBm) unit is defined with respect to power rather than voltage. This is commonly used in high frequency systems where one is interested in, for example, the power in an outward propagating wave generated by a source. In terms of power, it is defined as the average power over a cycle relative to 1 milliwatt:

$$
{\rm dBm} = 10 \log_{10} \left( \frac{P}{1\ {\rm mW}} \right)
$$

Note that following the Decibel convention, decibels are defined with a factor of 10 rather than a factor of 20. Sounds crazy, but it is very handy because if somebody tells you a value the amplitude transfer function in dB, you do not need to ask if they are talking about a "voltage ratio" or a "power ratio"! Takes some getting used to but it is handy.

For a voltage source supplying voltage $V_0 \cos(\omega t)$, the power it will actually deliver depends on what resistor you connect to it. However, when working with high frequency signals, for reasons we will see later in the course, it is common to be interested in the power delivered into a 50 ohm resistor, which then gives:

$$
{\rm dBm} = 10 \log_{10} 
\left( 
\frac{V_0^2 / (50\ {\rm \Omega}) / \sqrt{2}} 
{1\ {\rm mW}} 
\right)
$$

Here, the factor of $1/\sqrt{2}$ comes from the averaging over a cycle for the power dissipated in a resistor we derived in Lecture 1. This allows you to convert the voltage of your signal into units of dBm. 

## Complex representations of signals and impedances

In circuit analysis, as you saw above, it can be a bit awkward mathematically to work with sines and cosines, and for this reason, it is very common to instead work with a complex-valued [phasor](https://en.wikipedia.org/wiki/Phasor) representation of electrical signals. Doing so will make the math much easier and will allow us to use a powerful notation of complex impedance to greatly simplify calculations. 

### Phasor representation of signals

In [phasor](https://en.wikipedia.org/wiki/Phasor) notation, you represent oscillating signals by complex exponentials. Say I have a signal:

$$
V(t) = V_0 \cos (\omega t)
$$

The **phasor representation** of this signal is then:

$$
\tilde V(t) = \tilde V_0 e^{j \omega t}
$$

*(Note that in this course, we will follow the convention in electrical engineering of using the letter $j$ to represent $\sqrt{-1}$. This is because later in the course, we will use the small letter $i$ to represent small perturbations on a large current when performing [small signal analysis](https://en.wikipedia.org/wiki/Small-signal_model), and so it is [convenetion](https://en.wikipedia.org/wiki/Imaginary_number#Notes) to use the notation $j = \sqrt{-1}$.)*

Note that both $\tilde V(t)$ and $\tilde V_0$ are (potentailly) *complex valued* numbers. For now, we will put a tilde over all phasor variables just to be 100% clear which variables are a complex phasor. Later we will drop the tildes (just to save writing them all the time), but sometimes we will have to be careful: when you need to be careful is a something we will discuss at the end of the lectures. Although we use complex notation here as a mathematical tool, unlike the quantum wavefunction, the actual *real-life signals* are always real-valued: complex notation is just a trick we will use to make the mathematics easier. 

If you have a phasor, and you want to get to the real-valued signal, you need to take the real part:

$$
V(t) = {\rm Re} \left( \tilde V(t) \right)
$$

Using Euler's rule $e^{jx} = \cos(x) + j \sin(x)$, we can see that this will give us back our original $V(t)$. 

You can visualize the phasor as a vector in the complex plane that is rotating around a cirlce with an angular velocity given by $\omega$, as shown in this animation:

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
from matplotlib.patches import ConnectionPatch

# ---- parameters -------------------------------------------------------
A = 1.0                    # amplitude
phi0 = np.deg2rad(20)      # phase at t = 0 (static image)
n_periods = 2              # periods of the cosine shown on the right
color = "darkblue"
lim = 1.15 * A             # same y-range for BOTH panels

# ---- layout: left panel is square so 1 unit has the same size in x and y;
# right panel has the same height and shares the y-axis ------------------
fig_w, fig_h = 10, 3.6
pad, h = 0.03, 0.94
gap = 0.05                      # empty space between the two panels
w_p = h * fig_h / fig_w
plt.rcParams['animation.embed_limit'] = 50  # MB
fig = plt.figure(figsize=(fig_w, fig_h))
ax_p = fig.add_axes([pad, pad, w_p, h])
ax_s = fig.add_axes([pad + w_p + gap, pad, 1 - w_p - gap - 2 * pad, h], sharey=ax_p)
ax_p.set_xlim(-lim, lim)
ax_p.set_ylim(-lim, lim)
ax_s.set_xlim(-0.05, n_periods)

for ax in (ax_p, ax_s):
    ax.axhline(0, color="k", lw=0.8)
    ax.axvline(0, color="k", lw=0.8)
    ax.set_xticks([]); ax.set_yticks([])
    for s in ax.spines.values():
        s.set_visible(False)

# ---- static elements ---------------------------------------------------
th = np.linspace(0, 2 * np.pi, 400)
ax_p.plot(-A * np.sin(th), A * np.cos(th), ":", color=color, lw=2)  # circle
t = np.linspace(0, n_periods, 800)

# Labels rotated 90 deg counter-clockwise
ax_p.text(0, lim, "Re", rotation=90, ha="right", va="top", fontsize=13)
ax_p.text(lim, 0, "Im", rotation=90, ha="right", va="bottom", fontsize=13)
ax_p.text(-0.7*lim, 0.7*lim, r"$\tilde V(t)$", ha="right", va="bottom", fontsize=22, color=color)
ax_s.text(n_periods, 0, r"$t$", ha="right", va="bottom", fontsize=13)

# ---- dynamic elements --------------------------------------------------
phasor, = ax_p.plot([], [], "-", color=color, lw=2.5)
tip_dot, = ax_p.plot([], [], "o", color=color, ms=7)
wave, = ax_s.plot([], [], "-", color=color, lw=2.5)
sig_dot, = ax_s.plot([], [], "o", color=color, ms=7)
ylabel = ax_s.text(0, 0, r"$V(t)$", color=color, fontsize=13)
conn = ConnectionPatch((0, 0), (0, 0), coordsA=ax_p.transData,
                       coordsB=ax_s.transData, linestyle=":", color=color, lw=1.5)
fig.add_artist(conn)

def draw(theta):
    """theta = phase of the phasor; it increases counter-clockwise."""
    x, y = -A * np.sin(theta), A * np.cos(theta)    # (Im, Re): ccw from the +Re axis
    phasor.set_data([0, x], [0, y])
    tip_dot.set_data([x], [y])
    wave.set_data(t, A * np.cos(2 * np.pi * t + theta))
    sig_dot.set_data([0], [y])
    ylabel.set_position((0.04, y + 0.08 * A))
    conn.xy1, conn.xy2 = (x, y), (0, y)             # same y -> always horizontal

plt.close(fig)   # keep the figure object, but don't display it as a static image

n_frames = 72                                    # one full period
anim = FuncAnimation(fig, lambda k: draw(phi0 + 2 * np.pi * k / n_frames),
                     frames=n_frames)
html = anim.to_jshtml(fps=24, default_mode="loop")

css = f'''<style>
img[id^="_anim_img"] {{ width:800px; max-width:100%; height:auto }}
.anim-state {{ display:none }}        /* hides the Once / Loop / Reflect radio buttons */
.anim-slider {{ display:none }}
</style>'''
HTML(css + '<div style="display:flex; justify-content:center">' + html + '</div>')
```

As the phasor $\tilde V(t)$ rotates around in the complex plane, it's real part oscillates in the same way that our "real life" voltage signal $V(t)$ does. The actual physically meaningful signal $V(t)$ is then given by the projection of this vector onto the real axis, as shown in the animation.

What if our signal has a phase?

$$
V(t) = V_0 \cos(\omega t + \phi)
$$

In phasor notation, we will have:

$$
\tilde V(t) = V_0 e^{j(\omega t + \phi)} = V_0 e^{j\omega t} e^{j \phi} = \left(V_0 e^{j \phi} \right) e^{j \omega t}
$$

If we add a phase to our original signal, we can represent this in phasor notation as a **rotation** of the complex amplitude coefficient in the complex plane: 

$$
\tilde V_0 \rightarrow e^{j \phi} V_0
$$

Using the complex amplitude phasor $\tilde V_0$, we can now represent **both** the phase and the amplitude of the signal in one (complex-valued) number:

$$
{\rm Magnitude:} \ |\tilde V_0|
$$

$$
{\rm Phase:} \ \angle \tilde V_0
$$

The amplitude is given by the magnitude of the complex phasor, and the phase is given by the angle of the phasor in the complex plane.

### Solving the equations of motion with complex phasors

OK, so let's try solving the differential equations of the RC circuit for oscillating source voltages but now with complex phasor notation. Note that since the mathematical operation of taking the real part of a complex number is linear (${\rm Re}(z_1 + z_2)  {\rm Re}(z_1) + {\rm Re}(z_2)}, and because our equations of motion are linear differential equations, we are perfectly free to solve them first using phasors and then taking the real part of the solution: the real part of the phasor solution will also be equal to the solution of the linear differential equations of the real part. 

Our differential equation is:

$$
R \frac{dI}{dt} + \frac{I}{C} = \frac{dV_s}{dt}
$$

In phasor notation we will write:

$$
\tilde V_s(t) = \tilde V_s e^{j\omega t}
$$

$$
\tilde I(t) = \tilde I e^{j\omega t}
$$

where $\tilde V_s$ and $\tilde I$ are now time independent complex numbers. Plugging these into the equation, we get:

$$
j \omega R \tilde I e^{j \omega t}
+ \frac{\tilde I}{C} e^{j \omega t}
= j \omega \tilde V_s e^{j \omega t}
$$

We immediately from this see the advantage of complex notation: in taking the derivative, we do not switch from a cosine to a sine, and now, unlike before, we can immediately divide out the time dependence on both sides of the equation:

$$
j \omega R \tilde I 
+ \frac{\tilde I}{C}
= \left(j \omega R + \frac{1}{C} \right) \tilde I 
= j \omega \tilde V_s
$$

And now, look, we can solve this in one step:

$$
\tilde I 
= \frac{j \omega \tilde V_s}{j\omega R + \frac{1}{C}}
= \frac{j \omega C }{1 + j\omega RC}\ \tilde V_s
$$

Let's check if we get the same answer:

$$
|I|^2 = \tilde I^* \tilde I = \frac{\omega^2 C^2}{1 + \omega^2 R^2 C^2} |\tilde V_s|^2
$$

This gives an amplitude that matches our answer above derived with sines and cosines:

$$
|I| = \frac{\omega C V_s}{1 + \omega^2 R^2 C^2}  = I_0 
$$

We can also find the phase angle. First it is handy to rewrite things a bit:

$$
\tilde I 
= \frac{j \omega C }{1 + j\omega RC}\ \tilde V_s 
= \frac{1- j \omega RC}{1 + (\omega RC)^2}\ j \omega C \tilde V_s 
$$

From this we can now read off the phase angle:

$$
\tan \angle \tilde I = 
\frac{{\rm Im}(\tilde I)}{{\rm Re} (\tilde I)} = - \omega RC
$$

Up to a minus sign, this exactly matches the result we derived earlier. Why the sign difference? If you remember above, we chose the sign to be negative to make the math a bit cleaner. Here, we did not make that choice, and hence the sign associated with the lagging current shows up here explicitly. 

### Complex impedance

As you saw above, when working with phasor notation for theses oscillating signals, the equations we get from Kirchoff's laws are no longer differential equations, involving derivatives in time, but now instead algebraic equations with complex coefficient. Working with complex phasor notation make life much easier!

This allows us to introduce a different way of thinking capacitors and inductors by representing them by a [complex-valued impedance](https://en.wikipedia.org/wiki/Electrical_impedance#Complex_impedance). The idea is that we can write out an equivalent complex impedance $Z$ for all circuit elements that expresses the linear relations between the voltage and current phasors:

$$
\tilde V = \tilde I Z
$$

This looks very much like the relation $V = IR$ for a resistor, but now with phasors. To simplify our notation, from **now onwards in the course**, we will **drop the tildes** from our phasors:

$$
V = I Z
$$

You will have to determine yourself if the letters $V$ and $I$ are phasors or not from the context: if you see a letter $Z$ or any $j$ in the context, you know that we will be talking about phasors, and *most* of the time, we will work with phasor notation when discussing AC signals. 

*(One exception will come later in this lecture when we talk about power, and then later in the course, when we will breifly back to sines and cosines when talking about nonlinearities outside the context of small signal models. In these cases, we will be very explicit about what is a phasor and what is not and reintroduce the tildes so we are 100% clear.)*  

For a resistor, the complex impedance is purely real valued and give by $Z = R$

$$
Z_R = R
$$

For an inductor, using $V = L (dI / dt)$, we can see that taking the derivative will bring down a factor of $j \omega$, giving:

$$
Z_L = j \omega L
$$

For the capacitor, we can remember that $I = C (dV / dt)$, which, when we take the derivative and invert the relation, will give us:

$$
Z_C = \frac{1}{j \omega C}
$$

What is super useful about these complex impedances is that they also add in series and in parallel in the same way as resistors:

$$
Z_{\rm series} = \sum_i Z_i
$$

$$
\frac{1}
{Z_{\rm parllel}}
= \sum_i
\frac{1}
{Z_i}
$$

### Building intuition from frequency-dependent complex impedances

Not only does complex impedance make our math much easier, it also offers a nice way of intuitively understanding the frequency response of circuits.

Let's have another look at the RC filter again, along with the circuit but now redrawing the circuit so you can see it more directly as impedance divider:

```python
import base64, io
import numpy as np
import matplotlib.pyplot as plt
import schemdraw
import schemdraw.elements as elm
from IPython.display import HTML, display

HEIGHT_PX = 320  # common height for both figures

# ---------- Matplotlib plot ----------
norm_omega = np.logspace(-3, 3, 1000)
V_ratio = 1.0 / np.sqrt(1 + norm_omega ** 2)
dB = 20 * np.log10(V_ratio)

fig, ax1 = plt.subplots(1, 1, figsize=(5, 4))
ax1.semilogx(norm_omega, dB, color="b", linewidth=2)
ax1.set_xlabel(r"Normalized Frequency $\omega / \omega_c$")
ax1.set_ylabel(r"Relative Amplitude (dB)")
ax1.grid(True, which="both", linestyle="--", alpha=0.7)
fig.tight_layout()

buf = io.BytesIO()
fig.savefig(buf, format="svg", bbox_inches="tight", transparent=True)
plt.close(fig)  # prevent a second, separate display of the figure
plot_svg = buf.getvalue()

# ---------- Schemdraw drawing ----------
with schemdraw.Drawing(show=False) as d:
    elm.Line().length(1).right().label(r"$V_{\rm in}$", loc="left")
    elm.Dot()
    elm.ResistorIEC().label("$Z_R$").down()
    d.push()
    elm.Dot()
    elm.Line().length(1).right().label(r"$V_{\rm out}$", loc="right")
    d.pop()
    elm.ResistorIEC().down().label("$Z_C$")
    elm.Ground()

circuit_svg = d.get_imagedata("svg")

# ---------- Side-by-side, centered, same height ----------
HEIGHT_PX = 320        # height of the plot
CIRCUIT_SCALE = 0.85   # circuit height as a fraction of the plot height

def img_tag(svg_bytes, height_px):
    b64 = base64.b64encode(svg_bytes).decode("ascii")
    return (f'<img src="data:image/svg+xml;base64,{b64}" '
            f'style="height:{height_px}px; width:auto;">')

display(HTML(f"""
<div style="display:flex; justify-content:center; align-items:center;
            gap:8rem; width:100%;">
  {img_tag(plot_svg, HEIGHT_PX)}
  {img_tag(circuit_svg, int(HEIGHT_PX * CIRCUIT_SCALE))}
</div>
"""))
```

At low frequencies, the impedance of the capacitor becomes infinite:

$$
Z_C = \frac{1}{\omega C} \rightarrow \infty \ \ {\rm as} \ \ \omega \rightarrow 0
$$

If the impedance of the capacitor is infinite, then no current can flow: no voltage drops across the resistor and the full voltage of the input drops across the capacitor and appears directly at the output, giving a relative amplitude of 0 dB (voltage ratio of 1).  

At high frequencies, the impedance of the capacitor goes to zero:

$$
Z_C = \frac{1}{\omega C} \rightarrow 0 \ \ {\rm as} \ \ \omega \rightarrow \infty
$$

The capacitor becomes a short circuit and allows the high frequency current to flow freely with no voltage drop. All the voltage is the dropped across the resistor, and the output voltage then goes to zero: the relative amplitude goes to $-\infty$ dB as $\omega \rightarrow \infty$. 

Note here that when we talk about relative amplitude here using the decibel scale, we do not need to specify if we are talking about voltage amplitudes or power amplitudes, since the dB number will be the same for both. 

### Power with complex impedance and phasors

So far, we have seen that all linear behavior can be captured by working with complex-valued phasors and complex impedances and then taking the real part of our answers at the end. 

When one is interested in quantities that are *not* linear in voltage and current, one has to be a bit more careful. One nonlinear quantity we have already looked at is power:

$$
P(t) = I(t) V(t)
$$

For a resistor, with V = IR, we can see immediately that it is a nonlinear function of either current or voltage:

$$
P(t) = \frac{\left(V(t)\right)^2}{R} = \left( I(t) \right)^2 R
$$

What if we are working with phasors? Since we are now mixing time dependent voltages and phasors, we will put the tildes back on our phasors for now to be 100% sure which one we are talking about. If we have two phasors $\tilde V$ and $\tilde I$ representing a time dependent current and a time dependent voltage oscillating at a frequence $\omega$, one can show that the cycle averaged power as:

$$
P = \frac{1}{2} {\rm Re} \left( \tilde V \tilde I^* \right) 
$$

where $I^*$ is the complex conjugate of the current phasor. The factor of $\tfrac{1}{2}$ comes from the fact that we are working with peak amplitudes instead of RMS amplitudes. We can rewrite this in the following form:

$$
P = \frac{1}{2} |\tilde V| |\tilde I| \cos(\phi_V - \phi_I)
$$

It looks *almost* like we take the product of the current and the voltage, but there is another term from the cosine of the phase difference between the current and the voltage that also comes in. If we remember that currents and voltages in capacitors and inductors are offset by 90 degrees, we can see that this cosine will always be zero for a capacitor or an inductor, reflecting once again that capacitors and inductors themselves can never dissipate energy. On the other hand, for a resistor, currents and voltages are exactly in phase, and so this cosine factor becomes 1, and the resistor dissipates all the energy supplied to it. 

If you consider a current phasor $\tilde I$ flowing through a complex impedance $Z$, you can derive that the power dissipated is determined by the only the *real* part of the complex impedance:

$$
P = |\tilde I|^2\ {\rm Re}(Z)
$$

Since capacitors and inductors have complex impedances that are purely imaginary, we can see again that they will not disspate any power: the power dissipation in a circuit is related *only* to the real-valued part of its impedance. 

### Summary of phasors and complex impedance 

**Summary:** Phasor notation and complex impedence allow you to solve the frequency dependence of circuits containing resitors, inductors, and capacitors using (complex valued) *algebraic* equations using simple math! 

We can freely use this approach whenever we want, as long as we are a bit careful when we want to calculate any nonlinear quantities like power. 

For now, this new concpetual approach is limited to being able to predict the response of the circuit to a single frequency. However, later in the course when we study Fourier's theorem applied to electrical signals, we will discover that we can apply this *same* approach to study *any* time dependent signal by decomposing it in Fourier space. And doing so, we will be able to use algebraic approach and the intuition of complex impedance to understand and solve *any* time-dependent response of linear circuits.
