## Capacitors

Unlike a resistor, where the voltage is determined by the amount of current that is flowing, the voltage across a [capacitor](https://en.wikipedia.org/wiki/Capacitor) is determined by the *charge* on it's plates:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Dot().label("$V$")
    elm.Capacitor().down().label("C").scale(3)
    elm.Ground()
    elm.Label("+Q").at((0.7,-0.8))
    elm.Label("-Q").at((0.75,-2))
    elm.Label(r"$Q = CV$").at((3,-1.5))
    elm.Label(r"$V = \dfrac{Q}{C}$").at((5,-1.5))
    display_schematic(d)
```

One common type of capacitor is a [parallel plate capacitor](https://en.wikipedia.org/wiki/Capacitor#Parallel-plate_capacitor):

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Load the image
img = mpimg.imread("capacitor_field.png")

fig = plt.figure(dpi=600)
plt.imshow(img)
plt.axis("off")

show_plot(fig)
```

Most of the electric fields are concentrated in vertical lines between the two capacitor plates that carry the charge. Neglecting the fringing fields, we have a capacitance:

$$
C = \frac{\epsilon A}{d}
$$

where $A$ is the area of the plates, $d$ is the distance between the two plates, and $\epsilon$ is the dielectric constant of the material between the plates. 

Not only capacitors we put into our circuits have capacitance, but all cables and ever wire in our circuit will have a self-capacitance to each other and to ground, and to each other. The actual value of these capacitances depend strongly on gemoetry, but a good rule of thumb is that you can expect a self-capacitance of about 1 pF per cm of wire in your circuit, calculated by approximating the wire of lenght $\ell$ as a metallic sphere of diameter $\ell$ in free space and calculating the capacitance of that sphere to infinity ($C = 2 \pi \epsilon_0 \ell = 0.55$ pF / cm, giving the approximate order of magnitude of 1 pF / cm). 

As mentioned earlier, capacitors do not dissipate energy but instead store them in electric fields. The total energy stored in a capacitor is determined by the integral of the electric fields generated by the charges on the capacitor over all space, and can also be conveniently expressed as:

$$
E = \tfrac{1}{2} \int \epsilon |\vec E|^2 d^3x = \tfrac{1}{2} C V^2 = \frac{Q^2}{2C}
$$

By taking a derivative of the equation $Q = CV$, we can also find a relation between the current flowing into a capacitor and the voltage across it, which is useful if we want to use Kirchoff's Current Law:

```python
import schemdraw
import schemdraw.elements as elm

with schemdraw.Drawing(show=False) as d:
    elm.Dot().label("$V$")
    C = elm.Capacitor().down().label("C").scale(3)
    elm.Ground()
    elm.Label(r"$I = C\ \dfrac{dV}{dt}$").at((3,-1.5))
    elm.CurrentLabelInline(direction='in').at(C).label('I')
    display_schematic(d)
```

Note that if $V$ is time-independent, then the current through the capacitor is always zero: capacitors cannot transmit DC current. This property leads common application of a capacitor in a circuit as a "DC block" for voltages: it is a circuit element that can block DC voltages but allows AC voltages.

### Example: Kirchoff's laws with capacitors

We will consider an example with a voltage source in series with a capacitor and an inductor:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Ground()
    elm.Dot()
    S = elm.SourceV().up().label("$V_s(t)$")
    elm.CurrentLabelInline(direction='out', start=False).at(S).label('I')
    elm.Dot().label("$V_1$")
    elm.Resistor().label("R").right()
    elm.Dot().label("$V_2$")
    elm.Capacitor().down().label("C")
    elm.Line().left()
    display_schematic(d)
```

This current has one loop and zero branches. We will use Kirchoff's Votlage law around the loop of the circuit. 

$$
\sum \Delta V = 0 = (V_1 - 0) + (V_2 - V_1) + (0 - V_2)
$$

In the homework, you will look at a parallel RC circuit driven by a current source, for which you will use instead Kirchoff's Current Law. But back to the equation above, being careful with the signs, we can use $(V_2 - V_1) = - IR$ and $(0 - V_2) = -Q/C$ to get the following:

$$
0 = V_s(t) - I(t)R - \frac{Q(t)}{C}
$$

What should we do with $Q(t)$? This is given by the integral of $I(t)$ in time, but that would give us an integral equation (which is also fine, but we are likley more familiar with solving differential equations). We can also take the derivative of this equation to get a first order differential equation. Doing so and rearranging terms, we get:

$$
R \frac{dI}{dt} + \frac{I}{C} = \frac{dV_s}{dt}
$$

This is a nice familiar first order linear [ordinary differential equation](https://en.wikipedia.org/wiki/Ordinary_differential_equation) with a "forcing term" $dV_s / dt$. To be able to solve this differential equation, we first have to know what $V_0(t)$ is.

The example we will look at now is a specific case of $V_s(t)$ in the form of a step function:

$$
V_s(t) = 
\left\{ 
\begin{array}{ll} 
& 0 \hspace{0.5cm} & t <  0 \\
& V_0 \hspace{0.5cm} & t \gt 0 
\end{array}
\right.
$$

```python
mpl.rc('font', size=14)
fig = plt.figure(figsize=(7,3), dpi=100)
plt.axis('off')
plt.plot([-1.2,5.2],[0,0], 'k')
plt.plot([0,0],[-0.5,2.5], 'k')
plt.text(5, -0.4,"$t$")
plt.text(-0.5, 2.5,"$V_s$")
plt.text(-0.5, 1.9, "$V_0$")
t = np.linspace(-1,5, 1000)
V0 = (t>0)*2
plt.plot(t,V0, lw=3, c="blue")
show_plot(fig)
```

For $t<0$ and $t>0$, $dV_s / ldt$ is zero. Of course, something violent happens to $dV_s/dt$ exactly at $t=0$, but we will ignore that for now: we will first solve the differential equation for $t<0$ and $t>0$ and then think about how to apply the boundary conditions to match the solutions between the two regions. 

With $dV_s / dt$ = 0, our differential equation becomes:

$$
\frac{dI}{dt} = -\frac{1}{RC} I
$$

This one is likely familiar to you! The solution to this differential equation has the form:

$$
I = I_0 e^{-\gamma t}
$$

with $\gamma = 1/RC$. 

Now, let's look at the boundary conditions: if $V_s(t) = 0$ for all $t<0$, then the current in the resistor at all times $t<0$ will be zero, which means that the parameter $I_0$ = 0. Since currenet never flowed through the resistor, then the charge on the capacitor will also be zero. Thus the solution is $V_1(t<0) = V_2(t<0) = 0$. 

For $t>0$, we can ask: what happens to the current exactly at $t=0$ when the voltage jumps to $V_0$? 

Exactly when this jump happens, $V_2$ will not instantly react since charge needs to flow in order to build up voltage on the capacitor. With $V_2 = 0$, the current through the resistior at $t=0$ is then $I(t=0) = V_0 / R$, which allows us to determine $I_0 = V_0 / R$. For $t>0$, we then have:

$$
I(t) = \frac{V_0}{R} e^{-\gamma t}
$$

Finally, we can calculate $V_2(t)$ either by calculating the charge $Q(t) = \int I dt$, or we can also use $V_2 = V_s - IR$. Using the second, we can easily find:

$$
V_2(t) = V_0 \left( 1 - e^{-\gamma t} \right)
$$

```python
fig = plt.figure(figsize=(7,6), dpi=100)
plt.subplot(2,1,1)
plt.axis('off')
plt.plot([-1.2,5.2],[0,0], 'k')
plt.plot([0,0],[-0.5,2.5], 'k')
plt.text(5, -0.35,"$t$")
plt.text(-0.2, 2.5,"$I$")
plt.text(-0.9, 1.9, r"$I_0 = \frac{V_0}{R}$")
t = np.linspace(-1,5, 1000)
V0 = (t>0)*2
plt.plot([0,5.2],[V0,V0], lw=1, ls="--", c='k')
plt.plot(t,V0*np.exp(-t), c="red", lw=2)

# the 1/tau time
V_t1 = np.exp(-1)*2
plt.plot([1,1], [0, V_t1], ls=':', c='k')
plt.plot([0,1], [V_t1, V_t1], ls=':', c='k')
plt.text(-1.2, V_t1*0.9, r"$\approx (0.37) I_0$", fontsize = 12)
plt.text(0.92,-0.3, r"$\tau$", fontsize=12)



plt.subplot(2,1,2)
plt.axis('off')
plt.plot([-1.2,5.2],[0,0], 'k')
plt.plot([0,0],[-0.5,2.5], 'k')
plt.text(5, -0.35,"$t$")
plt.text(-0.4, 2.5,"$V_2$")
plt.text(-0.4, 1.9, "$V_0$")
plt.plot([0,5.2],[V0,V0], lw=1, ls="--", c='k')
plt.plot(t,V0*(1-np.exp(-t)), c="blue", lw=3)
# the 1/tau time
V_t1 = (1-np.exp(-1))*2
plt.plot([1,1], [0, V_t1], ls=':', c='k')
plt.plot([0,1], [V_t1, V_t1], ls=':', c='k')
plt.text(-1.7, V_t1*0.95, r"$\approx (1-0.37) V_0$", fontsize = 12)
plt.text(0.92,-0.3, r"$\tau$", fontsize=12)

show_plot(fig)
```

If you look at $I(t)$ in the upper plot, you can see that when there is a step in the source voltage $V_s(t)$, there is (in principle) an immediate jump of the current. The current flows through the resistor and starts charging up the capacitor: as the capacitor charges, the voltage $V_2$ builds up across the capacitor. As $V_2$ builds up, the voltage across the reistor drops, reducing the current until at long times, the capacitor is fully charged and the current goes to zero.

If you think about $V_s$ is as an input and $V_2$ as an output, then if you look at the output voltage $V_2$ across the capacitor, it takes some time for the jump in the input voltage $V_s$ to propagate to change in the output voltage $V_2$, and builds up over over a timescale $\tau = 1/\gamma$: in a time $t = \tau$, the output gets within a factor of $1/e \approx 37 \%$ of the final output at long times. When you think of the series RC circuit as a voltage low pass filter, then value $\tau$ is referred to as the "time constant" of the filter. 

## Inductors

The next circuit element we will look at is an [inductor](https://en.wikipedia.org/wiki/Inductor):

```python
with schemdraw.Drawing(show=False) as d:
    elm.Dot().label("$V$")
    L = elm.Inductor().down().label("L", ofst=0.3).scale(2)
    #elm.CurrentLabelInline(direction="in").at((0,-1.25)).label("$I$")
    elm.CurrentLabelInline(direction="in").at((0,-3.9)).label("$I$",ofst=(-0.1,0))
    elm.Ground()
    elm.Label(r"$V = L\ \dfrac{dI}{dT}$").at((2,-1.5))
    display_schematic(d)
```

An inductor is a bit of a strange thing: unlike a capacitors, it will allow a DC current to flow, and unlike resistors, an (ideal) inductor will allow a DC current to flow with *no* voltage drop!

However, if you want to *change* the current in an inductor, then you *do* need to supply a voltage, proportional to the rate at which you want to change the current:

$$
V = L\ \frac{dI}{dt}
$$

Why? The physical origin of the behavior of an inductor is in the magnetic fields that are created in and around the inductor when a current flows:

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Use a system LaTeX installation for all text rendering
plt.rcParams["text.usetex"] = True
plt.rcParams["font.family"] = "serif"

# Load the image
img = mpimg.imread("inductor_coil.png")

fig, (ax_img, ax_eq) = plt.subplots(
    1, 2,
    figsize=(7, 4),
    gridspec_kw={"width_ratios": [5, 1]},
)

# Left: image
ax_img.imshow(img)
ax_img.axis("off")

# Right: equation
ax_eq.text(
    0.5, 0.5,
    r"$\displaystyle E = \frac{1}{2} \int \mu |\vec B |^2 \  d^3x$",
    fontsize=24,
    ha="center",
    va="center",
)
ax_eq.axis("off")

show_plot(fig)

plt.rcParams["text.usetex"] = False
plt.rcParams["font.family"] = "sans-serif"
```

When current flows in the wires of the inductor, a magnetic field is created in the space around them. This magnetic field stores a energy given by the integral of square magnitude of the magnetic field over all space:

$$
E = \tfrac{1}{2} \int \mu | \vec B |^2 \ d^3x
$$

Changing the current in an inductor changes the energy stored in the magnetic fields around the inductor, and you therefore need a voltage to *change* the current in the inductor in order to increase this energy. The simplest inductor is a tighly wound coil like shown here:

```python
show_svg("solenoid_schematic.svg", scale=1)
```

As you see in the picture above, most of the magnetic field density is concentrated *inside* the coil. If we count only this, neglecting the the fringing fields, the inductance of an inductor is given by: 

$$
L_{\rm solonoid} = \mu n \pi r^2 \ell
$$

$n = N / \ell$ is the coil wrapping density in turns per unit length. For a fixed coil turn density, we can see that the inductance scales linearly with the length $\ell$ of the coil, and quadratically with the radius $r$: this is related to the volume of the magnetic field created by the inductor. $\mu = \mu_r \mu_0$ is the magnetic [permeability](https://en.wikipedia.org/wiki/Permeability_(electromagnetism)) of the volume inside the coil (the "core"), $\mu_0 = 1.25$ nH/mm, and $\mu_r$ is the relative (magnetic) permeability of the coil [core](https://en.wikipedia.org/wiki/Magnetic_core). Often a material with high magnetic permeability, such as soft iron ($\mu_r \approx 5,000 - 200,000$ at room temperature), is used to increase the coil inductance. In "high frequency" inductors, such materials are usually avoided due to losses in switching the magnetic domains at high frequencies, leading to an effective "resistance" of the inductor at high frequencies due to [core losses](https://en.wikipedia.org/wiki/Magnetic_core#Core_loss). 

Soloniod coils are not the only circuit element that has an inductance: *any* wire in your circuit will create magnetic fields and will therefore also have an associated inductance. Unlike the capacitance of a sphere to infinity, there is no closed form for the inductance of an isolated wire to infinity since it technically [diverges logarithmically](https://share.google/aimode/9vthF1t9vMvCFMwdD) as you take the current return path to infinity. However, in these formulas, the inductance still scales linearly iwth the wire length and a good practial "finger in the air" estimate for typical wires is an about 1 nH / mm.

### Example: Kirchoff's laws with inductors

We will consider the following circuit to explore how to apply Kirchoff's laws with inductors:

```python
with schemdraw.Drawing(show=False) as d:
    elm.Ground()
    elm.Dot()
    S = elm.SourceV().up().label("$V_s(t)$")
    elm.CurrentLabelInline(direction='out', start=False).at(S).label('I')
    elm.Dot().label("$V_1$")
    elm.Resistor().label("R").right()
    elm.Dot().label("$V_2$")
    elm.Inductor().down().label("L")
    elm.Line().left()
    display_schematic(d)
```

Again, we will apply Kirchoff's Voltage Law around the loop:

$$
\sum \Delta V = 0 = (V_1 - 0) + (V_2 - V_1) + (0 - V_2)
$$

Using $(V_2 - V_1) = - IR$ and $(0 - V_2) = -L dI/dt$ to get the following:

$$
0 = V_s(t) - I(t)R - L\frac{dI}{dt}
$$

which will give us the differential equation:

$$
L \frac{dI}{dt} + IR = V_s(t)
$$

Again, like the capacitor, we will look at the solutions of this equation for the case that $V_s(t)$ is a step function: 

$$
V_s(t) = 
\left\{ 
\begin{array}{ll} 
& 0 \hspace{0.5cm} & t <  0 \\
& V_0 \hspace{0.5cm} & t \gt 0 
\end{array}
\right.
$$

For $t<0$, we will get the equation:

$$
L \frac{dI}{dt} + IR = 0
$$

Applying the boundary condition that $V_s(t = -\infty) = 0$, we can see that for $t<0$ there will be no current and no voltage: $I(t<0) = V_2(t<0) = 0$. (There was never any voltage in the circuit and therefore it makes sense that there are not currents flowing.)

For $t>0$, we will have the following equation:

$$
L \frac{dI}{dt} + IR = V_0
$$

How can we solve this? One thing we can do is take the derivative of the equation in time:

$$
L \frac{d^2I}{dt^2} + R \frac{dI}{dt} = 0
$$

This is now a differential equation that is easier to solve: in fact, it looks identical in form to the one we solved for the capacitor but now with $I$ replaced by $dI/dt$. We could continue to solve this for $dI/dt$, but could also remember that $V_2 = L dI/dt$, which means we can rewrite this in terms of $V_2$:

$$
\frac{dV_2}{dt} + \frac{R}{L} V_2 = 0
$$

This is now *exactly* the differential equation we had above but for $V_2$ instead of $I$. We can then directly see that the solution is:

$$
V_2(t > 0) = V_{2}(t=0) e^{-\gamma t}
$$

with decay rate:

$$
\gamma = \frac{R}{L}
$$

and time constant $\tau = 1/\gamma = L/R$. The value $V_{2}(t=0)$ is a parameter that we need to find by applying the boundary conditions at $t=0$. 

What happens at $t=0$? When $V_s$ jumps up to $V_0$ at $t=0$, this will try to create spike in the current. However, unlike the capacitor, the inductor will not allow sudden jumps in current. Since it starts with no current flowing at $t=0$, it cannot suddenly increase it's current: a step function in the current through the inductor would require infinite voltage since $V = L dI/dt$. Instead what happens is that at short times, in response to this jump, the inductor looks like an infinite impedance: all the voltage $V_s$ drops across the inductor. Since no current is flowing in the resistor after the jump, we can see that $V_2(t=0) = V_s$. We then have:

$$
V_2(t > 0) = V_{2}(t=0) e^{-\gamma t}
$$

We can then also solve for the current using $I = (V_s - V_2(t))/R$, giving:

$$
I = I_0 (1-e^{-\gamma t})
$$

with $I_0 = V_s / R$, which is the steady-state current that the circuit comes to at long times. At long times, the current is no longer changing in time: the voltage drop across the inductor is zero, and the current is determined by only the resistor and the source voltage. 

We can also plot these, and contrast them to the case of the series RC circuit:

```python
fig = plt.figure(figsize=(14,7), dpi=100)

plt.subplot(2,2,2)
plt.axis('off')
plt.plot([-1.2,5.2],[0,0], 'k')
plt.plot([0,0],[-0.5,2.5], 'k')
plt.text(5, -0.3,"$t$")
plt.text(-0.2, 2.5,"$I$")
plt.text(-0.9, 1.9, r"$I_0 = \frac{V_0}{R}$")
t = np.linspace(-1,5, 1000)
V0 = (t>0)*2
plt.plot([0,5.2],[V0,V0], lw=1, ls="--", c='k')
plt.plot(t,V0*(1-np.exp(-t)), c="red", lw=2)
plt.title(r"Series RL circuit, $\tau = L/R$", pad=25)

plt.subplot(2,2,4)
plt.axis('off')
plt.plot([-1.2,5.2],[0,0], 'k')
plt.plot([0,0],[-0.5,2.5], 'k')
plt.text(5, -0.3,"$t$")
plt.text(-0.4, 2.5,"$V_2$")
plt.text(-0.4, 1.9, "$V_0$")
plt.plot([0,5.2],[V0,V0], lw=1, ls="--", c='k')
plt.plot(t,V0*np.exp(-t), c="blue", lw=3)


plt.subplot(2,2,1)
plt.axis('off')
plt.plot([-1.2,5.2],[0,0], 'k')
plt.plot([0,0],[-0.5,2.5], 'k')
plt.text(5, -0.3,"$t$")
plt.text(-0.2, 2.5,"$I$")
plt.text(-0.9, 1.9, r"$I_0 = \frac{V_0}{R}$")
t = np.linspace(-1,5, 1000)
V0 = (t>0)*2
plt.plot([0,5.2],[V0,V0], lw=1, ls="--", c='k')
plt.plot(t,V0*np.exp(-t), c="red", lw=2)
plt.title(r"Series RC circuit $\tau = RC$", pad=25)

plt.subplot(2,2,3)
plt.axis('off')
plt.plot([-1.2,5.2],[0,0], 'k')
plt.plot([0,0],[-0.5,2.5], 'k')
plt.text(5, -0.3,"$t$")
plt.text(-0.4, 2.5,"$V_2$")
plt.text(-0.4, 1.9, "$V_0$")
plt.plot([0,5.2],[V0,V0], lw=1, ls="--", c='k')
plt.plot(t,V0*(1-np.exp(-t)), c="blue", lw=3)

show_plot(fig, scale=0.9)
```

You can see that the roles of current and voltage are swapped: the RC circuit shows a spike in the current and then a slow relaxation of the voltage, while the RL circuit shows a spike in the voltage and a slow relaxation of the current. 

Like the series RC circuit, the series RL circuit can also be considered a filter that translates the input voltage $V_s(t)$ to an output voltage $V_2(t)$, but in this case, it is a high-pass filter instead of a low pass filter: it transfers "high frequency spikes" to the output tthrough but does not transfer the DC votlage from $V_s$ since the (ideal) inductor has no voltage drop for a constant current. 

In an RC circuit, if you want to increase the time constant $\tau$ (make the rise time slower), one way is to increase the capacitance: larter capacitors take longer to charge. The second way is to increase the resistance: a larger resistance lets less current through, so the capacitor charges more slosly. 

In an RL circuit, larger inductors also store more magnetic field energy: for larger inductors, the RL time constant $\tau$ also increases. What is perhaps not immediately obvious is that if you want a **larger** $RL$ time, you can also achieve this by **reducing** the size of the resistor! This was for me a bit counterintuitive: why does *decreasing* the resistance *increase* the RL time? 

The best way to understand this is to think about this as a competition between the resistor and the inductor. The inductor would like to do it's best to keep the current constant, while the resistor would like to let current flow in response to the input voltage. These two are in competetion for the voltage that is supplied. If the resistance is very small, then it will allow current to flow with almost no influence on the voltage that drops across it: in that case the behaviour of the circuit will be dominated by the inductance, which will generate a "back EMF" (thevoltage drop it creates across itself equal to $V_2(t)$) to resist the sudden jump in current. Left on it's own, with no resistance in the circuit, the inductor will sustain this back-EMF for ever! When we put the resistor in, we allow some of the voltage to begin to drop across the resistor, allowing the back-EMF $V_2(t)$ to eventually decay to zero at long times. The larger the resistor, the more quickly this voltage drop builds up across the resistor, and hence the more quickly the output voltage decays. 

In practice, since no voltage source has an ideal zero output resitance, and because the wires in our circuit and our inductor are also not zero resitance, in practice, the EMF generated by the inductor in response to the voltage step will always eventually decay to zero.
