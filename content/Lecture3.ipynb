# The differential equations of RC and RL circuits

> In this lecture, we will review the behaviour of capacitors and inductors in circuits, showing how to use Kirchoff's laws to derive differential equations that describe their behaviour in time. We will explore their solutions and provide physical intuition to the way they behave.


## Capacitors

Unlike a resistor, where the voltage is determined by the amount of current that is flowing, the voltage across a capacitor is determined by the *charge* on it's plates:

```python
import schemdraw
import schemdraw.elements as elm

with schemdraw.Drawing() as d:
    elm.Dot().label("$V$")
    elm.Capacitor().down().label("$C$").scale(3)
    elm.Ground()
    elm.Label("+Q").at((0.7,-0.8))
    elm.Label("-Q").at((0.75,-2))
    elm.Label(r"Q = CV").at((3,-1.5))
    elm.Label(r"$V = \dfrac{Q}{C}$").at((5,-1.5))

```

By taking a derivative of the equation for $Q$, we can also find a relation between the current flowing into a capacitor and the voltage across it:

```python
import schemdraw
import schemdraw.elements as elm

with schemdraw.Drawing() as d:
    elm.Dot().label("$V$")
    C = elm.Capacitor().down().label("$C$").scale(3)
    elm.Ground()
    elm.Label(r"$I = C\ \dfrac{dV}{dt}$").at((3,-1.5))
    elm.CurrentLabelInline(direction='in').at(C).label('I')
```
